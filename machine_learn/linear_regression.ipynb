线性回归模型
$$
f(x) = wx + b
\\ x: 输入 (feature)
\\ w: 权重 (weight)
\\ b: 偏置 (bias)
$$

当有多个特征时，我们可以使用多个权重来预测目标变量。
此时的线性回归模型为：
$$
f(x) = w_1x_1 + w_2x_2 + ... + w_nx_n + b
$$


```python
# 随机生成几组数据
import numpy as np
import matplotlib.pyplot as plt
# 生成100个样本，每个样本有2个特征
X = np.random.rand(100, 2)
# 生成对应的目标变量，每个样本的目标变量为两个特征的和加上一个随机噪声
y = np.dot(X, np.array([1.5, -2.0])) + np.random.randn(100)
```

```python
np.array([1.5, -2.0])
```

```python
plt.scatter(X[:, 0], y, label='Feature 1')
plt.scatter(X[:, 1], y, label='Feature 2')
plt.title('Scatter Plot of Features')
plt.xlabel('Feature Value')
plt.ylabel('Target Variable')
plt.legend()
plt.show()
```
