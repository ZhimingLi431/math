$$
\text{Sigmoid函数是一个常用的激活函数，它的数学表达式为：} \\ \quad \\
\sigma(x) = \frac{1}{1 + e^{-x}}
$$

$$
\text{Sigmoid的导数为：} \\ \quad \\
\sigma'(x) = \frac{e^{-x}}{(1 + e^{-x})^2} = \sigma(x) (1 - \sigma(x))
$$

$$
\text{Sigmoid函数的二阶导数为：} \\ \quad \\
\sigma''(x) = \sigma(x)'(1-2\sigma(x)) = \sigma(x)(1-\sigma(x))(1-2\sigma(x)) = \frac{e^{-x}(e^{-x}-1)}{(1+e^{-x})^{3}}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)
y = 1 / (1 + np.exp(-x))

# 计算Sigmoid函数的导数
y_prime = y * (1 - y)

# 计算Sigmoid函数的导数 (缺一个数据)
y_diff = [y[i + 1] - y[i] for i in range(len(y) - 1)]

# 计算Sigmoid函数的二阶导数
y_prime_prime = y * (1 - y) * (1 - 2 * y)

plt.scatter(x, y)
plt.plot(x, y_prime, color='red')
plt.plot(x[:-1], y_diff, color='green')
plt.plot(x, y_prime_prime, color='blue')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Sigmoid Function')
plt.legend(['Sigmoid', 'Derivative', 'Derivative (diff)', 'Second Derivative'])
plt.show()
```

```python
# 用 sigmoid 函数举一个激活函数的例子
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 生成一些随机数据
x = np.linspace(-10, 10, 100)
y = sigmoid(x)

# 绘制 sigmoid 函数的图像
plt.plot(x, y, label='Sigmoid Function')
plt.title('Sigmoid Function')
plt.xlabel('Input (x)')
plt.ylabel('Output (sigmoid(x))')
plt.axhline(0.5, color='r', linestyle='--', label='Threshold (0.5)')
plt.legend()
plt.show()
```

```python
# 假设我们有一个简单的线性模型，输出为 z = w1x1 + w2x2 + b
# 我们将 z 传递给 sigmoid 函数，得到预测的概率值
x1 = np.random.rand(100)
x2 = np.random.rand(100)
```

```python
w1 = 0.8
w2 = 0.6
b = -0.2
z = w1*x1 + w2*x2 + b
probability = sigmoid(z)
plt.figure(figsize=(12, 7))
plt.subplot(1, 2, 1)
plt.hist(probability, bins=20)
plt.title('Histogram of Predicted Probabilities')
plt.xlabel('Predicted Probability')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.scatter(x1, probability, label='Feature 1')
plt.scatter(x2, probability, label='Feature 2')
plt.title('Scatter Plot of Features vs Predicted Probabilities')
plt.xlabel('Feature Value')
plt.ylabel('Predicted Probability')
plt.legend()
plt.show()
```
