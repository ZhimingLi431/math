```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array(
    [
        [143, 88], [145, 85], [146, 88], [147, 91], [149, 92], [150, 93], [153, 93], 
        [154, 95], [155, 96], [156, 98], [157, 97], [158, 96], [159, 98], [160, 99], 
        [162, 100], [164, 102]
        ]
    )
x = data[:, 0]
y = data[:, 1]

plt.plot(x, y, 'o')
plt.xlabel('身高')
plt.ylabel('腿长')
plt.show()
```

$$
\text{一元线性回归模型 (Simple Linear Regression)} \\
\textbf{y = } \beta_0 + \beta_1 x + \epsilon \\
\quad \\
y \text{ 是模型的响应变量 (也称为因变量) (观测值)} \\
x \text{ 是模型的预测变量 (也称为自变量) (特征值)} \\
\beta_0 \text{ 是模型的截距项} \\
\beta_1 \text{ 是模型的系数项} \\
\epsilon \text{ 是模型的误差项} \\
$$

$$
一元线性回归方程 \\
E(y) = \beta_0 + \beta_1 x \\
\quad \\
E(y) \text{ 是方程的期望值} \\
\beta_0 \text{ 是方程的截距项} \\
\beta_1 \text{ 是方程的系数项} \\
$$

$$
残差 (Residual) \\
e_i = y_i - \hat{y}_i \\
\quad \\
e_i \text{ 是第 i 个观测值的残差} \\
y_i \text{ 是第 i 个观测值的响应变量 (也称为因变量) (观测值)} \\
\hat{y}_i \text{ 是第 i 个观测值的模型预测值 (也称为期望值)} \\
$$

$$
\text{推导过程}\\
\quad \\
\text{原方程} \\
y^{(i)} = \beta^T x^{(i)} + e^{(i)} \\
\quad \\
\text{正态分布表达式} \\
p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(x - \mu)^2}{2\sigma^2}} \\
\quad \\
\text{误差是独立分布, 并且服从正态分布 (高斯分布)} \\
p(e^{(i)}) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(e^{(i)})^2}{2\sigma^2}} \\
\quad \\
\text{将原方程代入} \\
p(y^{(i)} - \beta^T x^{(i)}) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(y^{(i)} - \beta^T x^{(i)})^2}{2\sigma^2}} \\
\quad \\
\text{似然函数: 对于独立同分布样本, 函数通常是各个观测值概率的乘积} \\
L(\beta) = L(x_1, x_2, \dots, x_n; \theta) =  \prod P(x_i; \theta) \\
\quad \\
\text{将概率方程代入似然函数} \\
L(\beta) = \prod_{i=1}^m P(y^{(i)} - \beta^T x^{(i)}; \beta) = \prod_{i=1}^m \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(y^{(i)} - \beta^T x^{(i)})^2}{2\sigma^2}} \\
\quad \\
\text{将乘法转换为加法} \\
\log L(\beta) = \log \prod_{i=1}^m \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(y^{(i)} - \beta^T x^{(i)})^2}{2\sigma^2}} \\
= \sum_{i=1}^m \log \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(y^{(i)} - \beta^T x^{(i)})^2}{2\sigma^2}} \\
= m \log \frac{1}{\sqrt{2\pi\sigma^2}} - \frac{1}{2\sigma^2} \sum_{i=1}^m (y^{(i)} - \beta^T x^{(i)})^2 \\
\quad \\
\text{要找到最大似然估计, 就是要找到使得似然函数值最大的参数值} \\
\quad \\
\text{最大似然估计 (Maximum Likelihood Estimation)} \\
\beta_{ML} = \arg\max_{\beta} L(\beta) \\
= \arg\max_{\beta} \log L(\beta) \\
= \arg\max_{\beta} \left( m \log \frac{1}{\sqrt{2\pi\sigma^2}} - \frac{1}{2\sigma^2} \sum_{i=1}^m (y^{(i)} - \beta^T x^{(i)})^2 \right)\\
= \arg\min_{\beta} \left( \frac{1}{2\sigma^2} \sum_{i=1}^m (y^{(i)} - \beta^T x^{(i)})^2 \right) \\
= \arg\min_{\beta} \left( \sum_{i=1}^m (y^{(i)} - \beta^T x^{(i)})^2 \right) \\
\quad \\
\text{推导结束}
$$

$$
\text{高斯 - 最小二层法 (Least Squares Method)} \\
\textbf{最小二层法是一种用于求解线性回归模型参数的方法, 其目标是最小化观测值与模型预测值之间的离差平方和} \\
\quad \\
\text{离差平方和 (Sum of Squared Deviation)} \\
Q = \sum (y - \hat{y})^2 = \sum_{i=1}^m (y - \beta_0 - \beta_1 x)^2 \\
= \sum (y - \beta_0 - \beta_1 x) (y - \beta_0 - \beta_1 x) \\
= \sum \left( y^2 - 2 y \beta_0 - 2 y \beta_1 x + \beta_0^2 + 2 \beta_0 \beta_1 x + \beta_1^2 x^2 \right) \\
= \sum y^2 - 2\beta_0 \sum y - 2 \beta_1 \sum yx + n\beta_0^2 + 2 \beta_0 \beta_1 \sum x + \beta_1^2 \sum x^2 \\
\quad \\
\text{求偏导, 并令其等于 0, 即可求解} \\
\frac{\partial Q}{\partial \beta_0} = -2 \sum y + 2 \beta_0 n + 2 \beta_1 \sum x = 0 \implies \sum y = \beta_0 n + \beta_1 \sum x \implies \beta_0 = \frac{\sum y - \beta_1 \sum x}{n} \implies \beta_0 = \bar y - \beta_1 \bar x \\
\quad \\
\frac{\partial Q}{\partial \beta_1} = -2 \sum yx + 2\beta_0 \sum x + 2 \beta_1 \sum x^2 = 0 \implies \sum yx = \beta_0 \sum x + \beta_1 \sum x^2 \cancel{\implies \beta_1 = \frac{\sum yx - \beta_0 \sum x}{n}} \\
\quad \\
\sum y = \beta_0 n + \beta_1 \sum x \implies \beta_0 = \frac{\sum y - \beta_1 \sum x}{n} \\
\sum yx = \beta_0 \sum x + \beta_1 \sum x^2 \implies \sum yx = \frac{\sum y \sum x - \beta_1 (\sum x)^2 + n \beta_1 \sum x^2}{n}
\quad \\
\implies n \sum yx = \sum y \sum x + \beta_1 (n\sum x^2 - (\sum x)^2) \implies \beta_1 = \frac{n \sum yx - \sum y \sum x}{n\sum x^2 - (\sum x)^2} \\
\quad \\
\beta_0 = \bar y - \beta_1 \bar x \\
$$

```python
sum_xy =  data.prod(axis=1).sum() # (x*y).sum()
sum_x, sum_y = data.sum(axis=0) # (x.sum(), y.sum())
sum_x2 = (data[ :, 0] ** 2).sum() # (x**2).sum()
x_bar = sum_x / x.size # x.mean()
y_bar = sum_y / y.size # y.mean()
n = data.shape[0]
print(f'sxy = {sum_xy}\nsy = {sum_y}\nsx = {sum_x}\nsx2 = {sum_x2}\nx_bar = {x_bar}\ny_bar = {y_bar}\nn = {n}')
```

```python
beta_1 = (n * sum_xy - sum_y * sum_x) / (n * sum_x2 - sum_x ** 2)
beta_0 = y_bar - beta_1 * x_bar

# 把 1 的列向量放到 x1 的前面, 代表截距项
A = np.vstack([np.ones(len(data[:, 0])), data[:, 0]]).T
coef, residuals, rank, s = np.linalg.lstsq(A, data[:, 1], rcond=None)

print(f'beta_1 = {beta_1}')
print(f'beta_0 = {beta_0}')
print(f'coef = {coef}')
```

$$
\text{估计标准误差 Standard Error of Estimate (SEE or } S_e \text{)} \\
\quad \\
S_e = \sqrt{\frac{\sum (y - \hat{y})^2}{n - k - 1}} \\
\text{其中, n 为样本数量, k 为自变量的数量, 1 为截距}
$$

```python
k = 1

def calc_standard_error_of_estimate(y, y_hat, n, k):
    return np.sqrt(((y - y_hat) ** 2).sum() / (n - k - 1))

S_e = calc_standard_error_of_estimate(data[ :, 1], y_hat, n, k)
print(f'S_e = {S_e}')
```

$$
\text{置信区间估计} \\
\quad \\
\text{对于每个预测值 \(\hat{y}_i\), 其置信区间为:} \\
\quad \\
\hat{y}_i \pm t_{\frac{\alpha}{2}} \times S_e \sqrt{\frac{1}{n} + \frac{(x_i - \bar{x})^2}{\sum (x - \bar{x})^2}} \\
\text{其中, \(t_{\frac{\alpha}{2}}\) 为 t 分布的 \(1-\alpha/2\) 分位数 }\\
\text{\(\alpha\) 为显著性水平 (significance level), \(1-\alpha\) 为置信水平 (confidence level), \(n\) 为样本数量, \(k\) 为自变量的数量}\\
\text{\(S_e\) 为估计标准误差, \(x_i\) 为第 \(i\) 个样本的预测值, \(\bar{x}\) 为所有样本的预测值的平均值} \\
$$

```python
import scipy.stats

alpha = 0.05

def calc_confidence_value(x_i, S_e, alpha=0.05, show_detail=False):
    t_alpha_2 = scipy.stats.t.ppf(1 - alpha/2, n - k - 1)
    sum_x_deviation = ((x - x_bar)**2).sum()
    confidence_value = t_alpha_2 * S_e * np.sqrt(1/n + (x_i - x_bar)**2 / sum_x_deviation)
    if show_detail:
        print(f't_alpha_2 = {t_alpha_2}')
        print(f'confidence_value = {confidence_value}')
    return confidence_value

def calc_confidence_interval(x_i, S_e, alpha=0.05, show_detail=False):
    confidence_value = calc_confidence_value(x_i, S_e, alpha, show_detail)
    lower_bound = y_hat - confidence_value
    upper_bound = y_hat + confidence_value
    if show_detail:
        print(f'lower_bound = {lower_bound}')
        print(f'upper_bound = {upper_bound}')
    return lower_bound, upper_bound

plt.scatter(x, y, label='Data')
plt.plot(x, y_hat, color='red', label='Fitted Line')
lower_bound, upper_bound = calc_confidence_interval(x, calc_standard_error_of_estimate(y, y_hat, n, k))
plt.fill_between(x, lower_bound, upper_bound, color='red', alpha=0.2, label='Confidence Interval of Fitted Line')
plt.legend()
plt.show()
```

$$
\text{预测区间估计} \\

\quad \\
\hat{y}_i \pm t_{\frac{\alpha}{2}} \times S_e \sqrt{1 + \frac{1}{n} + \frac{(x_i - \bar{x})^2}{\sum (x - \bar{x})^2}} \\
\text{其中, \(t_{\frac{\alpha}{2}}\) 为 t 分布的 \(1-\alpha/2\) 分位数 }\\
\text{\(\alpha\) 为显著性水平 (significance level), \(1-\alpha\) 为置信水平 (confidence level), \(n\) 为样本数量, \(k\) 为自变量的数量}\\
\text{\(S_e\) 为估计标准误差, \(x_i\) 为第 \(i\) 个样本的预测值, \(\bar{x}\) 为所有样本的预测值的平均值} \\
$$

```python
import scipy.stats

def calc_prediction_value(x_i, S_e, alpha=0.05, show_detail=False):
    t_alpha_2 = scipy.stats.t.ppf(1 - alpha/2, n - k - 1)
    sum_x_deviation = ((x - x_bar)**2).sum()
    prediction_value = t_alpha_2 * S_e * np.sqrt(1 + 1/n + (x_i - x_bar)**2 / sum_x_deviation)
    if show_detail:
        print(f't_alpha_2 = {t_alpha_2}')
        print(f'prediction_value = {prediction_value}')
    return prediction_value

def calc_prediction_interval(x_i, S_e, alpha=0.05, show_detail=False):
    prediction_value = calc_prediction_value(x_i, S_e, alpha, show_detail)
    lower_bound = y_hat - prediction_value
    upper_bound = y_hat + prediction_value
    if show_detail:
        print(f'lower_bound = {lower_bound}')
        print(f'upper_bound = {upper_bound}')
    return lower_bound, upper_bound

plt.scatter(x, y, label='Data')
plt.plot(x, y_hat, color='red', label='Fitted Line')
lower_bound, upper_bound = calc_prediction_interval(x, calc_standard_error_of_estimate(y, y_hat, n, k))
plt.fill_between(x, lower_bound, upper_bound, color='red', alpha=0.2, label='Prediction Interval of Fitted Line')
plt.legend()
plt.show()
```

$$
\text{回归直线的拟合优度 Goodness of Fit} \\
\quad \\
\text{TSS (Total Sum of Squares) (总平方和)} = \sum_{i=1}^n (y_i - \bar{y})^2 \\
\quad \\
\text{ESS (Explained Sum of Squares) (回归平方和) SSR (Sum of Squares Regression)} = \sum_{i=1}^n (\hat{y}_i - \bar{y})^2 \\
\quad \\
\text{RSS (Residual Sum of Squares) (残差平方和) SSE (Sum of Squares Error)} = \sum_{i=1}^n (y_i - \hat{y}_i)^2 \\
\quad \\
\text{TSS} = \text{ESS} + \text{RSS} \\
\quad \\
R^2 \text{( Coefficient of Determination) (决定系数)} = \frac{\text{ESS}}{\text{TSS}} = 1 - \frac{\text{RSS}}{\text{TSS}} \\
$$

```python
TSS = SST = ((y - y_bar) ** 2).sum()
ESS = SSR = ((y_hat - y_bar) ** 2).sum()
RSS = SSE = ((y - y_hat) ** 2).sum()
R2 = ESS / TSS
print(f'TSS = {TSS:.4f}')
print(f'ESS = {ESS:.4f}')
print(f'RSS = {RSS:.4f}')
print(f'R2 = {R2:.4f}')
```

$$
\text{显著性检验 (Significance Test)} \\
$$

$$
\text{线性关系检验 (Linear Relationship Test)} \\
\text{将均方回归(MSR) 同均方残差(MSE) 加以比较, 应用F检验来分析二者之间的差别是否显著} \\
\quad \\
\text{回归自由度 (Regression Degrees of Freedom, RDF): 回归模型中自变量的数量(k), 用于反映模型复杂度} \\
df_{Regression} = k \\
\quad \\
\text{误差自由度 (Error Degrees of Freedom, EDF): 回归模型中样本数量(n)-回归自由度(RDF)-1, 统计学中用于衡量回归模型中未被解释的变异部分的关键指标} \\
df_{Error} = n-k-1 \\
\quad \\
\text{均方回归 (Mean Squared Regression, MSR): 回归平方和(ESS)除以回归自由度(RDF)} \\
MSR = \dfrac{ESS}{df_{Regression}} \\
\quad \\
\text{均方残差 (Mean Squared Error, MSE): 残差平方和(RSS)除以误差自由度(EDF)} \\
MSE = \dfrac{RSS}{df_{Error}} \\
$$

$$
\text{原假设 H0: 线性回归模型的系数为0, 即不存在线性关系} \\
\text{备择假设 H1: 线性回归模型的系数不为0, 即存在线性关系} \\
\quad \\
\text{假设检验统计量: F-统计量} \\
\text{假设检验统计量的分布: F-分布} \\
\quad \\
\text{H0 成立时, } F = \dfrac{ESS / 1}{RSS / (n - 1 - 1)} = \frac{MSR}{MSE} \sim F(1, n - 2) \\
\quad \\
\text{如果 } F > F_{1-\alpha}(1, n - 2) \text{, 则拒绝 H0, 即存在线性关系} \\
$$

```python
df_regression = k
df_error = n - k - 1
MSR = ESS / df_regression
MSE = RSS / df_error
F = MSR / MSE
print(f'df_regression = {df_regression}')
print(f'df_error = {df_error}')
print(f'MSR = {MSR:.4f}')
print(f'MSE = {MSE:.4f}')
print(f'F = {F:.4f}')
```

```python
f_distribution = scipy.stats.f(df_regression, df_error)
f_critical = f_distribution.ppf(1 - alpha)
print(f'F({df_regression}, {df_error}) = {f_critical:.4f}')
```

```python
# plot f_distribution
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 4, 1000)
y = f_distribution.pdf(x)
# 绘制图像
plt.figure(figsize=(3, 2))
plt.plot(x, y, label=f'F({df_regression}, {df_error})')
plt.xlabel('x')
plt.ylabel('PDF')
plt.title('F-Distribution')
plt.legend()
plt.show()
```

```python
if F > f_critical:
    print('拒绝H0假设')
else:
    print('接受H0假设')
```

$$
\text{回归系数的显著性检验} \\
\quad \\
H0: \beta_j = 0: \text{Y 不依赖于 }X_j \\
H1: \beta_j \neq 0: \text{Y 依赖于 }X_j \\
\quad \\
\hat{\beta}_1 \text{ 是根据最小二乘法求出的样本统计量, 服从正态分布} \\
\quad \\
\hat{\beta}_1 \text{ 的分布具有如下性质} \\
\quad \\
\text{数学期望: } E(\hat{\beta}_1) = \beta_1 \\
\quad \\
\text{标准差: } \sigma_{\hat{\beta}_1} = \frac{\sigma}{\sqrt{\sum x_i^2 - \frac{1}{n}(\sum x_i)^2}} \\
\text{由于} \sigma \text{未知, 所以使用其估计量} S_e \text{来代替得到的} \hat{\beta_1} \text{的估计标准差} \\
\quad \\
S_{\hat{\beta}_1} = \frac{S_e}{\sqrt{\sum x_i^2 - \frac{1}{n}(\sum x_i)^2}} \\
\quad \\
\text{剩余标准差 (Residual Standard Error):} \\
S_e = \sqrt{\frac{1}{n-k-1} \sum_{i=1}^n (y_i - \hat{y}_i)^2} = \sqrt{MSE}\\
\quad \\
\text{t检验的统计量: } t = \frac{\hat{\beta}_1 - \beta_1}{S_{\hat{\beta}_1}} \sim t_{n-k-1} \\


```python
S_hat_beta_1 = S_e / np.sqrt(sum(x**2) - (1/n) * sum(x**2))
S_hat_beta_1
```

```python
t = (beta_1 - 0) / S_hat_beta_1
t_crit = scipy.stats.t.ppf(1-alpha/2, n-k-1)
if abs(t) > t_crit:
    print('拒绝 H0')
else:
    print('接受 H0')
```

$$
\text{多元回归分析 Multivariate Linear Regression} \\
\text{回归平面方程} \\
y = \beta_0 + \beta_1x_1 + \beta_2x_2 + \cdots + \beta_kx_k \\
\quad \\
\text{多元线性回归预测模型一般式} \\
y = \beta_0 + \beta_1x_1 + \beta_2x_2 + \cdots + \beta_kx_k + \epsilon \\
Y = X\beta + \epsilon \\
E(\epsilon) = 0 \text{, } COV(\epsilon, \epsilon) = \sigma^2 I_n \\
\text{k元线性回归模型, 并简记为} (Y, X\beta, \sigma^2 I_n) \\
\quad \\
Y = \begin{bmatrix}
y_1 \\
y_2 \\
\vdots \\
y_n
\end{bmatrix} , X = \begin{bmatrix}
1 & x_{11} & x_{12} & \cdots & x_{1k} \\
1 & x_{21} & x_{22} & \cdots & x_{2k} \\
\vdots & \vdots & \vdots & \ddots & \vdots \\
1 & x_{n1} & x_{n2} & \cdots & x_{nk}
\end{bmatrix} , \beta = \begin{bmatrix}
\beta_0 \\
\beta_1 \\
\vdots \\
\beta_k
\end{bmatrix} , \epsilon = \begin{bmatrix}
\epsilon_1 \\
\epsilon_2 \\
\vdots \\
\epsilon_n
\end{bmatrix} \\
\quad \\
$$

$$
\text{多元线性回归模型的参数估计} \\
Q = \sum_{i=1}^n \epsilon_i^2 = \sum_{i=1}^n (y_i - \beta_0 - \beta_1x_{i1} - \beta_2x_{i2} - \cdots - \beta_kx_{ik})^2 \\ \quad \\
\text{选择} \beta_0, \beta_1, \beta_2, \cdots, \beta_k \text{使} Q \text{最小, 也就是误差平方和} \sum_{i=1}^n \epsilon_i^2 \text{最小} \\ \quad \\
\text{根据矩阵内积的性质, 可以将Q写成矩阵形式} \\
Q = \epsilon^T \epsilon = (Y - X\beta)^T(Y - X\beta) \\
= (Y^T - \beta^TX^T)(Y - X\beta) = Y^TY - Y^TX\beta - \beta^TX^TY + \beta^TX^TX\beta \\
\text{这里第二项和第三项是相等的} \\
Y\text{(n x 1); }X\text{(n x k); } \beta\text{(k x 1)} \\
Y^T\text{(1 x n); }X^T\text{(k x n); } \beta^T\text{(1 x k)} \\ \quad \\
Y^TX\beta \implies \text{(1 x n) (n x k) (k x 1), 是一个标量} \\ \quad \\
\text{对 }\beta^TX^TY \text{取转置} \implies (\beta^TX^TY)^T = Y^T(X^T)^T(\beta^T)^T = Y^TX\beta \\
\text{最终结果} \\
Y^TY - Y^TX\beta - \beta^TX^TY + \beta^TX^TX\beta = Y^TY - 2\beta^TX^TY + \beta^TX^TX\beta \\ \quad \\
\text{对} \beta \text{求导, 并令导数等于0, 即可得到参数估计的解析解} \\
\frac{\partial Q}{\partial \beta} = \frac{\partial (Y^TY - 2\beta^TX^TY + \beta^TX^TX\beta)}{\partial \beta} \\ \quad \\
\frac{\partial(Y^TY)}{\partial \beta} = 0 \\ \quad \\
\frac{\partial(2\beta^TX^TY)}{\partial \beta}\text{ 令} a = 2X^TY \text{ 则} \\ \quad \\
\beta^Ta = \beta_1a_1 + \beta_2a_2 + \cdots + \beta_ka_k \implies \frac{\partial(\beta_1a_1 + \beta_2a_2 + \cdots + \beta_ka_k)}{\partial \beta} = a_1 + a_2 + \cdots + a_k = a = 2X^TY \\ \quad \\
\frac{\partial(\beta^TX^TX\beta)}{\partial \beta} \text { 令} A = X^TX \text{ (对称矩阵) 则} \\ \quad \\
\beta^TA\beta = 
\begin{bmatrix} \beta_1 & \beta_2 \end{bmatrix}
\begin{bmatrix}
A_{11} & A_{12} \\
A_{21} & A_{22} \\
\end{bmatrix}
\begin{bmatrix} \beta_1 \\ \beta_2 \end{bmatrix}
= A_{11}\beta_1^2 + A_{12}\beta_1\beta_2 + A_{21}\beta_1\beta_2 + A_{22}\beta_2^2 \\ \quad \\
= -2X^TY + 2X^TX\beta = 0 \\ \quad \\
\frac{\partial(\beta^TA\beta)}{\partial \beta} = 
\frac{\partial(A_{11}\beta_1^2 + A_{12}\beta_1\beta_2 + A_{21}\beta_1\beta_2 + A_{22}\beta_2^2)}{\partial \beta_1} \\
= 2A_{11}\beta_1 + A_{12}\beta_2 + A_{21}\beta_2 \\
\text{因为A是对称矩阵, 所以} A_{12} = A_{21} \\
= 2A_{11}\beta_1 + 2A_{12}\beta_2 \\
\text{这个就是 2A} \beta \text{的第1个元素, 以此类推} \\ \quad \\
\frac{\partial(\beta^TA\beta)}{\partial \beta} = 2A\beta = 2X^TX\beta \\ \quad \\
\frac{\partial Q}{\partial \beta} = \frac{\partial (Y^TY - 2\beta^TX^TY + \beta^TX^TX\beta)}{\partial \beta} \\
=\frac{\partial(Y^TY)}{\partial \beta} - \frac{\partial(2\beta^TX^TY)}{\partial \beta} + \frac{\partial(\beta^TX^TX\beta)}{\partial \beta} \\
= 0 - 2X^TY + 2X^TX\beta \\ \quad \\
= -2X^TY + 2X^TX\beta = 0 \implies X^TX\beta = X^TY \\ \quad \\
\implies \beta = (X^TX)^{-1}X^TY \\ \quad \\
$$

$$
\text{把 } \beta \text{代入回归平面方程得到: } \\
\quad \\
y = \hat{\beta_0} + \hat{\beta_1}x_1 + \hat{\beta_2}x_2 + \cdots + \hat{\beta_k}x_k \\
\text{经验回归平面方程} \\
$$

$$
\text{多元线性回归模型的检验} \\
\quad \\
\text{F检验法} \\ \quad \\
\text{(H0) 假设模型中的所有系数都是0, } F = \frac{U/k}{Q_e/(n-k-1)} \sim F(k, n-k-1) \\ \quad \\
\text{如果} F > F_{k, n-k-1}(1-\alpha) \text{, 则拒绝H0} \\
\text{认为模型中至少有一个系数不是0, 认为y与x有显著的线性关系} \\
\text{否则接受H0, 认为y与x没有显著的线性关系} \\ \quad \\
U = \sum_{i=1}^n (\hat{y_i} - \bar{y})^2 \\
Q_e = \sum_{i=1}^n (y_i - \hat{y}_i)^2 \\
k = \text{模型中自变量的个数} \\ \quad \\

\text{R检验法} \\ \quad \\
R = \sqrt{\frac{U}{L_{yy}}} = \sqrt{\frac{U}{U + Q_e}} \\
\text{R 称为y 与 x的多元相关系数或复相关系数, 取值范围为[0, 1]} \\ \quad \\

F = \frac{n - k - 1}{k} \frac{R^2}{1 - R^2} \\ \quad \\

\text{F 和 R检验等效}
$$

```python
# 每一行代表：[推销开支 x1, 实际账目数 x2, 同类商品竞争数 x3, 地区销售潜力 x4, 建材销售量 y]
data = np.array([
    [5.5, 31, 10, 8, 79.3], [2.5, 55, 8, 6, 200.1], [8.0, 67, 12, 9, 163.2], [3.0, 50, 7, 16, 200.1], [3.0, 38, 8, 15, 146.0],
    [2.9, 71, 12, 17, 177.7], [8.0, 30, 12, 8, 30.9], [9.0, 56, 5, 10, 291.9], [4.0, 42, 8, 4, 160.0], [6.5, 73, 5, 16, 339.4],
    [5.5, 60, 11, 7, 159.6], [5.0, 44, 12, 12, 86.3], [6.0, 50, 6, 6, 237.5], [5.0, 39, 10, 4, 107.2], [3.5, 55, 10, 4, 155.0],
    [8.0, 70, 6, 14, 201.4], [6.0, 40, 11, 6, 100.2], [4.0, 50, 11, 8, 135.8], [7.5, 62, 9, 13, 223.3], [7.0, 59, 9, 11, 195.0]
])

x0 = np.ones(len(data[:, 0]))
x1 = data[:, 0]
x2 = data[:, 1]
x3 = data[:, 2]
x4 = data[:, 3]
y = data[:, 4]
```

```python
# 1. 运算：结果是一个包含 (系数向量, 残差, 秩, 奇异值) 的元组
results, residuals, rank, s = np.linalg.lstsq(np.vstack((x0, x1, x2, x3, x4)).T, y, rcond=None)

# 2. 提取：
intercept = results[0]  # 因为 np.ones 在第一列
c1, c2, c3, c4 = results[1:]

print(f'(用Numpy库) y = {intercept:.4f} + {c1:.4f}x1 + {c2:.4f}x2 + {c3:.4f}x3 + {c4:.4f}x4')
```

```python
import statsmodels.api as sm

# 1. 整理数据 (X 需要包含常数项，这里假设你的 x0 已经是全 1 向量)
X = np.column_stack((x0, x1, x2, x3, x4))

# 2. 核心代码：一行拟合并获取结果
model = sm.OLS(y, X).fit()

# 3. 输出完整的统计报表（包含置信区间、p值等所有信息）
print(model.summary())

# 如果只要看参数的置信区间（默认 95%）
# print("\n系数置信区间: \n", model.conf_int())

# 如果要获取新数据 x_new 的预测区间
# pred = model.get_prediction(x_new)
# print(pred.summary_frame(alpha=0.05))



# 1. 获取模型原始预测值（用于画预测散点）
y_pred_samples = model.fittedvalues

def plot_with_prediction_points(model, X, y, feature_idx, feature_name):
    # --- 构造背景区间（固定其他变量为均值） ---
    n_points = 100
    x_mean = np.mean(X, axis=0)
    x_fake = np.tile(x_mean, (n_points, 1))
    x_range = np.linspace(X[:, feature_idx].min(), X[:, feature_idx].max(), n_points)
    x_fake[:, feature_idx] = x_range
    
    res = model.get_prediction(x_fake).summary_frame(alpha=0.05)
    
    # --- 绘图 ---
    # 1. 绘制区间
    plt.fill_between(x_range, res['obs_ci_lower'], res['obs_ci_upper'], color='red', alpha=0.2, label='95% predict interval')
    plt.fill_between(x_range, res['mean_ci_lower'], res['mean_ci_upper'], color='blue', alpha=0.2, label='95% confidence interval')
    plt.plot(x_range, res['mean'], color='blue', lw=1.5, label='Mean Fit')

    # 2. 绘制原始数据散点 (蓝色圆点)
    plt.scatter(X[:, feature_idx], y, color='gray', s=15, alpha=0.4, label='Raw Data')

    # 3. 核心：绘制模型生成的【预测点】 (红色叉号)
    # 这些点反映了模型在考虑了所有变量后，对当前样本的判断
    plt.scatter(X[:, feature_idx], y_pred_samples, color='red', marker='x', s=25, alpha=0.7, label='Model Predictions')
    
    plt.title(f'Analysis of {feature_name}')
    plt.xlabel(feature_name)
    plt.ylabel('y')
    plt.legend(fontsize='small')

# 布局显示
plt.figure(figsize=(8, 6))
names = ['Intercept', 'x1', 'x2', 'x3', 'x4']
for i in range(1, 5):
    plt.subplot(2, 2, i)
    plot_with_prediction_points(model, X, y, i, names[i])

plt.tight_layout()
plt.show()

```

```python
X = np.vstack((x0, x1, x2, x3, x4)).T
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print(f'(用公式计算) y = {intercept:.4f} + {c1:.4f}x1 + {c2:.4f}x2 + {c3:.4f}x3 + {c4:.4f}x4')

y_pred = X @ beta

plt.figure(figsize=(6, 5))

# 定义一个绘图辅助函数，避免重复代码
def plot_sub(index, xi, name):
    plt.subplot(2, 2, index)
    # 画出原始观测数据
    plt.scatter(xi, y, alpha=0.5, label='Raw data')
    # 画出模型预测值（散点形式，避免连线错乱）
    plt.scatter(xi, y_pred, color='red', marker='x', s=20, label='Predicted')
    plt.xlabel(name)
    plt.ylabel('y')
    plt.title(f'Relation: {name} vs y')
    plt.legend()

plot_sub(1, x1, 'x1')
plot_sub(2, x2, 'x2')
plot_sub(3, x3, 'x3')
plot_sub(4, x4, 'x4')

plt.tight_layout()
plt.show()
```

```python
import scipy.stats as stats
# 统计量
n, k = X.shape
k -= 1
dof = n - k - 1
alpha = 0.05
t_crit = stats.t.ppf(1 - alpha/2, dof)

# calculate MSE
residuals = y - y_pred
mse = np.sum(residuals**2) / dof
S_e = np.sqrt(mse)

# calculate (X^T X)^-1
xtx_inv = np.linalg.inv(X.T @ X)

# --- 为绘图准备区间数据 ---
# 为了画出平滑的区间带，我们需要构造“平滑的 X_fake” (固定其他变量为均值)
def get_intervals(feature_idx):
    x_range = np.linspace(X[:, feature_idx].min(), X[:, feature_idx].max(), 100)
    x_mean_val = np.mean(X, axis=0)
    
    # 构造测试矩阵：除了目标列，其余全是均值
    X_test = np.tile(x_mean_val, (100, 1))
    X_test[:, feature_idx] = x_range
    
    y_test_pred = X_test @ beta
    
    # 手动计算每个点的标准误差
    # CI SE: sqrt(x_i' * (X'X)^-1 * x_i * mse)
    # PI SE: sqrt((1 + x_i' * (X'X)^-1 * x_i) * mse)
    ci_margin = []
    pi_margin = []
    for i in range(100):
        xi = X_test[i, :]
        # 计算杠杆值：x_i' * (X'X)^-1 * x_i
        leverage = xi @ xtx_inv @ xi.T
        ci_margin.append(t_crit * S_e * np.sqrt(leverage))
        pi_margin.append(t_crit * S_e * np.sqrt(1 + leverage))
        
    return x_range, y_test_pred, np.array(ci_margin), np.array(pi_margin)
```

```python
plt.figure(figsize=(8, 6))

def plot_sub_with_intervals(index, xi, name, feature_idx):
    plt.subplot(2, 2, index)
    
    # 获取区间带数据
    x_range, y_line, ci, pi = get_intervals(feature_idx)
    
    # 1. 画区间带
    plt.fill_between(x_range, y_line - pi, y_line + pi, color='red', alpha=0.2, label='95% PI')
    plt.fill_between(x_range, y_line - ci, y_line + ci, color='blue', alpha=0.2, label='95% CI')
    plt.plot(x_range, y_line, color='blue', lw=1, label='Mean Trend')

    # 2. 画原始散点和预测点
    plt.scatter(xi, y, alpha=0.4, s=15, label='Raw data', color='gray')
    plt.scatter(xi, y_pred, color='red', marker='x', s=20, label='Predicted')
    
    plt.title(f'{name} Analysis')
    plt.legend(fontsize='small')

# 依次画出 x1, x2, x3, x4 (索引从 1 开始，因为 0 是截距项)
plot_sub_with_intervals(1, x1, 'x1', 1)
plot_sub_with_intervals(2, x2, 'x2', 2)
plot_sub_with_intervals(3, x3, 'x3', 3)
plot_sub_with_intervals(4, x4, 'x4', 4)

plt.tight_layout()
plt.show()
```
