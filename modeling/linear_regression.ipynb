```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array(
    [
        [143, 88], [145, 85], [146, 88], [147, 91], [149, 92], [150, 93], [153, 93], 
        [154, 95], [155, 96], [156, 98], [157, 97], [158, 96], [159, 98], [160, 99], 
        [162, 100], [164, 102]
        ]
    )

plt.plot(data[:, 0], data[:, 1], 'o')
plt.xlabel('身高')
plt.ylabel('腿长')
plt.show()
```

$$
一元线性回归模型 \\
\textbf{y = } \beta_0 + \beta_1 x + \epsilon \\
\quad \\
y \text{ 是模型的响应变量 (也称为因变量) (观测值)} \\
x \text{ 是模型的预测变量 (也称为自变量) (特征值)} \\
\beta_0 \text{ 是模型的截距项} \\
\beta_1 \text{ 是模型的系数项} \\
\epsilon \text{ 是模型的误差项} \\
$$

$$
一元线性回归方程 \\
E(y) = \beta_0 + \beta_1 x \\
\quad \\
E(y) \text{ 是方程的期望值} \\
\beta_0 \text{ 是方程的截距项} \\
\beta_1 \text{ 是方程的系数项} \\
$$

$$
残差 (Residual) \\
e_i = y_i - \hat{y}_i \\
\quad \\
e_i \text{ 是第 i 个观测值的残差} \\
y_i \text{ 是第 i 个观测值的响应变量 (也称为因变量) (观测值)} \\
\hat{y}_i \text{ 是第 i 个观测值的模型预测值 (也称为期望值)} \\
$$

$$
\text{高斯 - 最小二层法 (Least Squares Method)} \\
\textbf{最小二层法是一种用于求解线性回归模型参数的方法, 其目标是最小化观测值与模型预测值之间的离差平方和} \\
\quad \\
\text{离差平方和 (Sum of Squared Deviation)} \\
Q = \sum (y - \hat{y})^2 = \sum_{i=1}^m (y - \beta_0 - \beta_1 x)^2 \\
= \sum (y - \beta_0 - \beta_1 x) (y - \beta_0 - \beta_1 x) \\
= \sum \left( y^2 - 2 y \beta_0 - 2 y \beta_1 x + \beta_0^2 + 2 \beta_0 \beta_1 x + \beta_1^2 x^2 \right) \\
= \sum y^2 - 2\beta_0 \sum y - 2 \beta_1 \sum yx + n\beta_0^2 + 2 \beta_0 \beta_1 \sum x + \beta_1^2 \sum x^2 \\
\quad \\
\text{求偏导} \\
\frac{\partial Q}{\partial \beta_0} = -2 \sum y + 2 \beta_0 n + 2 \beta_1 \sum x = 0 \implies \beta_0 = \frac{\sum y - \beta_1 \sum x}{n} \\
\quad \\
\frac{\partial Q}{\partial \beta_1} = -2 \sum yx + 2 \beta_1 \sum x^2 = 0 \implies \beta_1 = \frac{\sum yx - \beta_0 \sum x}{n} \\
\quad \\
$$

$$
\text{推导过程}\\
\quad \\
\text{原方程} \\
y^{(i)} = \beta^T x^{(i)} + e^{(i)} \\
\quad \\
\text{正态分布表达式} \\
p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(x - \mu)^2}{2\sigma^2}} \\
\quad \\
\text{误差是独立分布, 并且服从正态分布 (高斯分布)} \\
p(e^{(i)}) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(e^{(i)})^2}{2\sigma^2}} \\
\quad \\
\text{将原方程代入} \\
p(y^{(i)} - \beta^T x^{(i)}) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(y^{(i)} - \beta^T x^{(i)})^2}{2\sigma^2}} \\
\quad \\
\text{似然函数: 对于独立同分布样本, 函数通常是各个观测值概率的乘积} \\
L(\beta) = L(x_1, x_2, \dots, x_n; \theta) =  \prod P(x_i; \theta) \\
\quad \\
\text{将概率方程代入似然函数} \\
L(\beta) = \prod_{i=1}^m P(y^{(i)} - \beta^T x^{(i)}; \beta) = \prod_{i=1}^m \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(y^{(i)} - \beta^T x^{(i)})^2}{2\sigma^2}} \\
\quad \\
\text{将乘法转换为加法} \\
\log L(\beta) = \log \prod_{i=1}^m \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(y^{(i)} - \beta^T x^{(i)})^2}{2\sigma^2}} \\
= \sum_{i=1}^m \log \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(y^{(i)} - \beta^T x^{(i)})^2}{2\sigma^2}} \\
= m \log \frac{1}{\sqrt{2\pi\sigma^2}} - \frac{1}{2\sigma^2} \sum_{i=1}^m (y^{(i)} - \beta^T x^{(i)})^2 \\
\quad \\
\text{要找到最大似然估计, 就是要找到使得似然函数值最大的参数值} \\
\quad \\
\text{最大似然估计 (Maximum Likelihood Estimation)} \\
\beta_{ML} = \arg\max_{\beta} L(\beta) \\
= \arg\max_{\beta} \log L(\beta) \\
= \arg\max_{\beta} \left( m \log \frac{1}{\sqrt{2\pi\sigma^2}} - \frac{1}{2\sigma^2} \sum_{i=1}^m (y^{(i)} - \beta^T x^{(i)})^2 \right)\\
= \arg\min_{\beta} \left( \frac{1}{2\sigma^2} \sum_{i=1}^m (y^{(i)} - \beta^T x^{(i)})^2 \right) \\
= \arg\min_{\beta} \left( \sum_{i=1}^m (y^{(i)} - \beta^T x^{(i)})^2 \right) \\
\quad \\
\text{推导结束}
$$
