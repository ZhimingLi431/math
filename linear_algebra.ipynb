$$
\text{矩阵的秩 (Rank) r(A): 指的是有效信息的行. 经过初等行变换之后, 剩下的个数就为这个矩阵的秩} \\
$$


```python
import numpy as np
import scipy


matrix = np.array([[1, 2, 3, 4, 5], [11, 12, 13, 14, 15], [222, 333, 442, 244, 133], [2541, 555, 553, 777, 888], [999, 642, 777, 666, 555]])
print(f'matrix rank = {np.linalg.matrix_rank(matrix)}')
```

$$
矩阵的零度 (Nullity): 指的是矩阵的解空间的维数 \\
和秩的关系：\\
\text{Rank(A) + Nullity(A) = n} \\
\text{其中，n 是矩阵的列数} \\
$$


```python
matrix = np.array([[1, 2, 3, 4, 5], [11, 12, 13, 14, 15], [111, 112, 113, 114, 115], [1000, 2000, 3000, 4000, 5000], [4, 42, 77, 66, 55]])
print(f'matrix nullity = {len(matrix[0]) - np.linalg.matrix_rank(matrix)}')
```

$$
\text{矩阵的迹 (Trace) tr(A): 指方阵对角线上所有元素之和} \\
tr(A) = \sum_{i=1}^n A_{ii} \\
\text{迹等于所有特征值之和: } tr(A) = \sum_{i=1}^n \lambda_i \\
\quad \text{证明思路: } \det(\lambda I - A) = (\lambda - \lambda_1)(\lambda - \lambda_2)\cdots(\lambda - \lambda_n) \text{, 比较两边 } \lambda^{n-1} \text{ 的系数} \\
\quad \text{左边展开后 } \lambda^{n-1} \text{ 只来自对角线的乘积 } (\lambda - a_{11})\cdots(\lambda - a_{nn}) \text{, 系数为 } -(a_{11} + \cdots + a_{nn}) \text{, 右边为 } -(\lambda_1 + \cdots + \lambda_n) \\
\text{当A矩阵的秩为1时, 零度为n-1, 也就是说 } A\textbf{x} = 0 \text{ 有 n-1 个线性无关的解, 特征值 0 至少出现 n-1 次, 最多只剩一个非零特征值 } \lambda \text{, 那么} \\
tr(A) = \lambda + 0 + 0 + ... + 0 \\
tr(A) = \lambda \\
\text{其中} \lambda \text{为A的一个特征值} \\
$$


```python
matrix = np.array([[1, 2, 3, 4, 5], [11, 12, 13, 14, 15], [111, 112, 113, 114, 115], [1000, 2000, 3000, 4000, 5000], [4, 42, 77, 66, 55]])
print(f'matrix trace = {np.trace(matrix)}')
print(f'特征值之和 = {np.sum(np.linalg.eigvals(matrix)).real:.6f}')

# 秩为 1 的矩阵 u v^T: 唯一的非零特征值 = v^T u = 迹
u = np.array([[1], [2], [3]])
v = np.array([[4, 5, 6]])
rank_1 = u @ v
print(f'rank_1 = \n{rank_1}')
print(f'秩 = {np.linalg.matrix_rank(rank_1)}, 特征值 = {np.round(np.linalg.eigvals(rank_1).real, 6)}, 迹 = {np.trace(rank_1)}, v^T u = {(v @ u).item()}')
```

$$
\text{矩阵的行列式 (Determinant): 指的是一个方阵的特征值的乘积, 可以用于判断矩阵是否可逆, 以及缩放比例} \\
det(A) = GaussianElimination(A) -> \text{最终得到的上三角矩阵的对角线元素的乘积 (每做一次行交换, 结果要乘以 -1)} \\
det(A) \neq 0 \text{ 时, 矩阵可逆, 面积 (体积) 的缩放比例为 } |det(A)| \text{, det(A) 为负表示空间方向发生了翻转 (镜像)} \\
det(A) = 0 \text{ 时, 矩阵变换后, 会导致降维, 继而不可逆} \\
$$


```python
matrix = np.array([[1, 2, 3, 4, 5], [11, 12, 13, 14, 15], [222, 333, 442, 244, 133], [2541, 555, 553, 777, 888], [999, 642, 777, 666, 555]])
print(f'matrix determinant = {np.linalg.det(matrix)}')
```

$$

\text{Gauss-Jordan 消元法 (初等变换): 是一种用于求解线性方程组的方法, 同时也可以用于求矩阵的逆} \\
[A|I] = [I|A^{-1}] \\
例子: \\
[A|I] = \begin{bmatrix}

0 & 0 & 1 & -5 & 4 & 1 \\
\end{bmatrix} \\
$$

$$
矩阵乘法的数学通式 \\
C_{ij} = \sum_{k=1}^{n} A_{ik}B_{kj} \\
C = \begin{bmatrix}
    \sum_{k=1}^{n} A_{1k}B_{k1} & \sum_{k=1}^{n} A_{1k}B_{k2} & \cdots & \sum_{k=1}^{n} A_{1k}B_{kn} \\
    \sum_{k=1}^{n} A_{2k}B_{k1} & \sum_{k=1}^{n} A_{2k}B_{k2} & \cdots & \sum_{k=1}^{n} A_{2k}B_{kn} \\
    \vdots & \vdots & \ddots & \vdots \\
    \sum_{k=1}^{n} A_{nk}B_{k1} & \sum_{k=1}^{n} A_{nk}B_{k2} & \cdots & \sum_{k=1}^{n} A_{nk}B_{kn} \\
\end{bmatrix} \\
如果B是列向量 \\
C = \begin{bmatrix}
    \sum_{k=1}^{n} A_{1k}B_{k1} \\
    \sum_{k=1}^{n} A_{2k}B_{k1} \\
    \vdots \\
    \sum_{k=1}^{n} A_{nk}B_{k1} \\
\end{bmatrix} \\
$$

```python
A = np.array([[1, 2, 3], [0, 1, 4], [5, 6, 0]])
B = np.array([[1, 0], [5, 2], [1, 6]])
C = A @ B
print(f'C = \n{C}')
```

$$
\text{矩阵的转置 (Transpose): 指的是将矩阵的行和列互换得到的新矩阵} \\
A_{ij}^T = A_{ji} \\
\quad \\
(A + Bx)^T = A^T + (Bx)^T = A^T + x^TB^T \\
(A + BCx)^T = A^T + (BCx)^T = A^T + x^TC^TB^T \\
(AB)^T = B^TA^T \\
$$


```python
matrix = np.array([[1, 2, 3], [0, 1, 4], [5, 6, 0]])
print(f'A^T = \n{matrix.T}')

matrix_2 = np.array([[11, 12, 13], [10, 11, 14], [15, 16, 10]])
X = np.array([[3, 4, 5], [6, 7, 8], [9, 1, 2]])
print(f'(A+Bx)^T == A^T + x^T B^T? {np.array_equal((matrix + matrix_2 @ X).T, (matrix.T + X.T @ matrix_2.T))}')
print(f'(AB)^T == B^T A^T? {np.array_equal((matrix @ matrix_2).T, (matrix_2.T @ matrix.T))}')
```

$$
\text{矩阵的特征值 (Eigenvalue): 指的是在矩阵变换下, 特征向量方向保持不变 (或正好反向), 长度发生改变的比例} \\
\textbf{Ax} = \lambda I \textbf{x} \\
(\textbf{A} - \lambda I)\textbf{x} = 0 \\
\text{其中, A 是矩阵, x 是特征向量, I 是单位矩阵} \\
\text{对于Mx=0, 有两种方法, 一是M矩阵可逆, 那x只有唯一解0, 要使x非零解, 那M矩阵必须是不可逆也就是奇异矩阵} \\
\text{不可逆的充分必要条件即是行列式为0} \\
\det(\textbf{A} - \lambda I) = 0 \\
\lambda\text{ 是特征值, 要找的就是这个缩放后的}\lambda \\

\text{矩阵的特征向量 (Eigenvector): 指的是在矩阵变换下, 方向保持不变的向量} \\
\text{将求得的特征值代入到 }(\textbf{A} - \lambda I)\textbf{x} = 0 \text{中, 求得x即可得到对应的特征向量} \\
$$


```python
matrix = np.array([[1, 2, 3], [0, 1, 4], [5, 6, 0]])
eig = np.linalg.eig(matrix)
print(f'matrix eigenvalues = {eig[0]}')
print(f'matrix eigenvectors = \n{eig[1]}')
```

$$
\text{对称矩阵 Symmetric Matrix} \\
\quad \\
A = \begin{bmatrix}
1 & 2 & 3 & 4 \\
2 & 5 & 6 & 7 \\
3 & 6 & 8 & 9 \\
4 & 7 & 9 & 10 \\
\end{bmatrix} = A^T \\
\quad \\
\text{对称矩阵的性质} \\
\text{1. 对称矩阵的元素满足} a_{ij} = a_{ji} \text{ (对角线元素没有正负的限制)} \\
\text{2. 对称矩阵的特征值都是实数 (但不一定非负, 特征值都非负的对称矩阵叫半正定矩阵, 见后面的正定矩阵)} \\
\text{3. 对称矩阵不同特征值对应的特征向量是正交的} \\
\quad \text{证明: } A\textbf{x} = \lambda_1 \textbf{x}, A\textbf{y} = \lambda_2 \textbf{y}, \lambda_1 \neq \lambda_2 \\
\quad \lambda_1 \textbf{x}^T\textbf{y} = (A\textbf{x})^T\textbf{y} = \textbf{x}^TA^T\textbf{y} = \textbf{x}^TA\textbf{y} = \lambda_2 \textbf{x}^T\textbf{y} \implies (\lambda_1 - \lambda_2)\textbf{x}^T\textbf{y} = 0 \implies \textbf{x}^T\textbf{y} = 0 \\
\text{4. 对称矩阵一定可以正交对角化: } A = Q \Lambda Q^T \text{, Q 是正交矩阵} \\
\text{下面例子里的矩阵就有一个负的特征值} \\
\quad \\
$$

```python
A = np.array([[1, 2, 3, 4], [2, 5, 6, 7], [3, 6, 8, 9], [4, 7, 9, 10]])
eig = np.linalg.eig(A)
print(f'symmetric matrix = \n{A}')
print(f'symmetric matrix eigenvalues = {eig[0]}')
print(f'symmetric matrix eigenvectors = \n{eig[1]}')
print(f'向量1和向量2是否正交? {eig[1][:, 0] @ eig[1][:, 1].T}')
print(f'向量1和向量3是否正交? {eig[1][:, 0] @ eig[1][:, 2].T}')
print(f'向量1和向量4是否正交? {eig[1][:, 0] @ eig[1][:, 3].T}')
print(f'向量2和向量3是否正交? {eig[1][:, 1] @ eig[1][:, 2].T}')
print(f'向量2和向量4是否正交? {eig[1][:, 1] @ eig[1][:, 3].T}')
print(f'向量3和向量4是否正交? {eig[1][:, 2] @ eig[1][:, 3].T}')
```

$$
\text{矩阵的对角化 (Diagonalization): 指的是将一个矩阵表示为特征向量矩阵和对角矩阵的乘积, 其中每个特征向量都是一个列向量} \\
A = PDP^{-1} \\
\text{其中, P 是特征向量矩阵, D 是特征值矩阵} \\ \quad \\
\text{推导: 设 A 有 n 个线性无关的特征向量 } \textbf{p}_1, \cdots, \textbf{p}_n \text{, 对应特征值 } \lambda_1, \cdots, \lambda_n \\
A\textbf{p}_i = \lambda_i \textbf{p}_i \implies A[\textbf{p}_1, \cdots, \textbf{p}_n] = [\lambda_1\textbf{p}_1, \cdots, \lambda_n\textbf{p}_n] = [\textbf{p}_1, \cdots, \textbf{p}_n]
\begin{bmatrix}
\lambda_1 & & \\
& \ddots & \\
& & \lambda_n
\end{bmatrix} \\
\implies AP = PD \implies A = PDP^{-1} \quad \text{(P 的列线性无关, 所以 P 可逆)} \\ \quad \\
\text{可对角化的条件: A 有 n 个线性无关的特征向量 (有 n 个不同的特征值是充分条件, 但不是必要条件)} \\
\text{不能对角化的例子: } \begin{bmatrix} 1 & 1 \\ 0 & 1 \end{bmatrix} \text{ 特征值 1 重复两次, 但只有一个线性无关的特征向量} \\
\text{对称矩阵一定可以对角化, 而且 P 可以选成正交矩阵 Q, 记作 } A = Q \Lambda Q^T \\ \quad \\
\text{应用: 计算矩阵的高次幂} \\
A^k = (PDP^{-1})(PDP^{-1})\cdots(PDP^{-1}) = PD(P^{-1}P)D(P^{-1}P)\cdots DP^{-1} = PD^kP^{-1} \\
D^k = \begin{bmatrix}
\lambda_1^k & & \\
& \ddots & \\
& & \lambda_n^k
\end{bmatrix} \text{ 只需要对角线元素各自求 k 次方} \\
$$

```python
A = np.array([[4, 1], [2, 3]])
eigenvalues, P = np.linalg.eig(A)
D = np.diag(eigenvalues)
P_inv = np.linalg.inv(P)
print(f'特征值 = {eigenvalues}')
print(f'P (特征向量按列排列) = \n{P}')
print(f'D = \n{D}')
print(f'P D P^-1 == A ? {np.allclose(P @ D @ P_inv, A)}')

# 应用: A^k = P D^k P^-1
k = 10
A_k_diag = P @ np.diag(eigenvalues ** k) @ P_inv
A_k_direct = np.linalg.matrix_power(A, k)
print(f'\nA^{k} (对角化) = \n{A_k_diag}')
print(f'A^{k} (直接相乘) = \n{A_k_direct}')
print(f'是否相等? {np.allclose(A_k_diag, A_k_direct)}')

# 不能对角化的例子: 特征向量矩阵 P 的秩不满, 不可逆
J = np.array([[1, 1], [0, 1]])
eigenvalues_J, P_J = np.linalg.eig(J)
print(f'\nJ 的特征值 = {eigenvalues_J}')
print(f'J 的特征向量矩阵 P = \n{P_J}')
print(f'P 的秩 = {np.linalg.matrix_rank(P_J)} (小于 2, 所以 J 不能对角化)')
```

$$
\text{幂零矩阵 (Nilpotent Matrix): 指一个方阵A, 满足} A^k = 0 \text{, 其中k为一个正整数} \\
\text{严格上(下)三角矩阵：主对角线及其以下(以上)元素全为0的矩阵, n 阶严格上三角矩阵满足 } N^n = 0 \text{, 所以一定是幂零矩阵} \\
\begin{bmatrix}
    0 & a_{12} & \cdots & a_{1n} \\
    0 & 0 & \cdots & a_{2n} \\
    \vdots & \vdots & \ddots & \vdots \\
    0 & 0 & \cdots & 0 \\
\end{bmatrix}^n = 
\begin{bmatrix}
    0 & 0 & \cdots & 0 \\
    0 & 0 & \cdots & 0 \\
    \vdots & \vdots & \ddots & \vdots \\
    0 & 0 & \cdots & 0 \\
\end{bmatrix} \\ \quad \\
\text{幂零矩阵的性质} \\
\text{1. 特征值全部为 0} \\
\quad \text{证明: 设 } N\textbf{v} = \lambda \textbf{v} \text{, 则 } N^k\textbf{v} = \lambda^k \textbf{v} \text{. 因为 } N^k = 0 \text{, 所以 } \lambda^k \textbf{v} = 0 \text{, 而 } \textbf{v} \neq 0 \text{, 所以 } \lambda = 0 \\
\text{2. 迹和行列式都为 0: } tr(N) = \sum \lambda_i = 0, \quad \det(N) = \prod \lambda_i = 0 \\
\text{3. } I - N \text{ 一定可逆, 而且逆矩阵是一个有限项的几何级数 (类比 } \frac{1}{1-x} = 1 + x + x^2 + \cdots \text{)} \\
(I - N)^{-1} = I + N + N^2 + \cdots + N^{k-1} \\
\quad \text{证明: } (I - N)(I + N + \cdots + N^{k-1}) = I + N + \cdots + N^{k-1} - N - N^2 - \cdots - N^k = I - N^k = I \\
$$

```python
# 严格上三角矩阵: 每乘一次, 非零元素就往右上角退一条对角线
N = np.array([[0, 2, 3, 4],
              [0, 0, 5, 6],
              [0, 0, 0, 7],
              [0, 0, 0, 0]])
for k in range(1, 5):
    print(f'N^{k} = \n{np.linalg.matrix_power(N, k)}\n')

print(f'N 的特征值 = {np.linalg.eigvals(N)}')
print(f'tr(N) = {np.trace(N)}, det(N) = {np.linalg.det(N)}')

# (I - N)^-1 = I + N + N^2 + N^3
I = np.eye(4)
geometric_series = I + N + np.linalg.matrix_power(N, 2) + np.linalg.matrix_power(N, 3)
print(f'(I - N)^-1 == I + N + N^2 + N^3 ? {np.allclose(np.linalg.inv(I - N), geometric_series)}')
```

$$
\text{正定矩阵 (Positive Definite Matrix)} \\
\text{对称矩阵 A, 如果对任意非零向量 } \textbf{x} \text{ 都有 } \textbf{x}^TA\textbf{x} > 0 \text{, 就叫正定矩阵; 如果是 } \ge 0 \text{, 叫半正定矩阵 (Positive Semi-Definite)} \\ \quad \\
\text{等价条件 (满足任意一条即可): } \\
\text{1. 所有特征值 } \lambda_i > 0 \\
\quad \text{证明: 对称矩阵 } A = Q\Lambda Q^T \text{, 令 } \textbf{y} = Q^T\textbf{x} \text{, 则 } \textbf{x}^TA\textbf{x} = \textbf{y}^T\Lambda\textbf{y} = \sum_i \lambda_i y_i^2 \text{, 只有全部 } \lambda_i > 0 \text{ 时才对所有非零 y 都大于 0} \\
\text{2. 所有顺序主子式 (左上角 k x k 子矩阵的行列式) 都 > 0 (Sylvester 判据)} \\
\text{3. 存在 Cholesky 分解 } A = LL^T \text{, L 是对角线元素为正的下三角矩阵} \\
\quad L_{jj} = \sqrt{a_{jj} - \sum_{k<j} L_{jk}^2}, \quad L_{ij} = \frac{1}{L_{jj}}\left(a_{ij} - \sum_{k<j} L_{ik}L_{jk}\right) \ (i > j) \text{ (根号里出现负数, 就说明不是正定)} \\ \quad \\
\text{几何意义: } f(\textbf{x}) = \textbf{x}^TA\textbf{x} \text{ 是一个 "碗" 形的二次函数, 只有一个最低点, 所以正定和最优化密切相关} \\
\text{例如最小二乘的 Hessian 矩阵 } \frac{\partial^2 Q}{\partial \beta \partial \beta^T} = 2X^TX \text{ 是半正定的, 所以导数为 0 的点一定是最小值} \\ \quad \\
\text{常见的 (半) 正定矩阵: 协方差矩阵, } X^TX \text{ (X 列满秩时正定)} \\
\quad \text{证明: } \textbf{v}^TX^TX\textbf{v} = (X\textbf{v})^T(X\textbf{v}) = \|X\textbf{v}\|^2 \ge 0 \\ \quad \\
\text{回到 statistics 里自协方差的问题: 为什么分母用 n 能保证自协方差矩阵是 (半) 正定的?} \\
\text{记中心化后的序列 } z_t = x_t - \bar{x} \text{ (t 超出 1..n 的部分补 0), 有偏估计 } \hat{\gamma}_k = \frac{1}{n}\sum_t z_t z_{t+k} \\
\textbf{a}^T\hat{\Gamma}\textbf{a} = \sum_{i,j} a_i a_j \hat{\gamma}_{|i-j|} = \frac{1}{n}\sum_t \left(\sum_j a_j z_{t+j}\right)^2 \ge 0 \\
\text{分母用 n-k 时, 每个 } \hat{\gamma}_k \text{ 除以的数不一样, 这个 "平方和" 的结构被破坏, 就可能出现负的特征值}
$$

```python
A_pd = np.array([[4, 2, 0],
                 [2, 5, 3],
                 [0, 3, 6]], dtype=float)

print(f'特征值 = {np.linalg.eigvalsh(A_pd)} (全部 > 0)')
print(f'顺序主子式 = {[round(float(np.linalg.det(A_pd[:k, :k])), 4) for k in range(1, 4)]} (全部 > 0)')

def cholesky_manual(A):
    n = A.shape[0]
    L = np.zeros_like(A, dtype=float)
    for i in range(n):
        for j in range(i + 1):
            s = A[i, j] - L[i, :j] @ L[j, :j]
            if i == j:
                if s <= 0:
                    raise ValueError('不是正定矩阵')
                L[i, j] = np.sqrt(s)
            else:
                L[i, j] = s / L[j, j]
    return L

L = cholesky_manual(A_pd)
print(f'L = \n{L}')
print(f'L L^T == A ? {np.allclose(L @ L.T, A_pd)}, 和 np.linalg.cholesky 一致? {np.allclose(L, np.linalg.cholesky(A_pd))}')

rng = np.random.default_rng(0)
random_x = rng.normal(size=(10000, 3))
quadratic_forms = np.einsum('ij,jk,ik->i', random_x, A_pd, random_x)   # 每一行算 x^T A x
print(f'10000 个随机向量的 x^T A x 最小值 = {quadratic_forms.min():.4f} (> 0)')

# 不是正定的对称矩阵 (前面对称矩阵的例子, 有一个负特征值)
A_sym = np.array([[1, 2, 3, 4], [2, 5, 6, 7], [3, 6, 8, 9], [4, 7, 9, 10]], dtype=float)
try:
    cholesky_manual(A_sym)
except ValueError as e:
    print(f'\n对称但不正定的矩阵: {e}, 特征值 = {np.round(np.linalg.eigvalsh(A_sym), 4)}')

# 自协方差矩阵: 有偏 (分母 n) vs 无偏 (分母 n-k)
from scipy.linalg import toeplitz
x_ts = np.array([1, 2, 6, 7, 2, 5, 7, 9, 4, 8, 2, 9, 1, 5, 6], dtype=float)
n_ts = len(x_ts)
z_ts = x_ts - x_ts.mean()
gamma_biased = [np.sum(z_ts[:n_ts - k] * z_ts[k:]) / n_ts for k in range(n_ts)]
gamma_unbiased = [np.sum(z_ts[:n_ts - k] * z_ts[k:]) / (n_ts - k) for k in range(n_ts)]
print(f'\n自协方差矩阵 (分母 n)   的最小特征值 = {np.linalg.eigvalsh(toeplitz(gamma_biased)).min():.4f}')
print(f'自协方差矩阵 (分母 n-k) 的最小特征值 = {np.linalg.eigvalsh(toeplitz(gamma_unbiased)).min():.4f} (< 0, 不是半正定)')
```

$$
\text{矩阵求导 (Matrix Calculus) 常用公式} \\
\text{约定: 标量 f 对列向量 } \textbf{x} \text{ 求导, 结果是同样形状的列向量 (梯度) } \nabla f = \left[\frac{\partial f}{\partial x_1}, \cdots, \frac{\partial f}{\partial x_n}\right]^T \\ \quad \\
\text{1. } \frac{\partial (\textbf{a}^T\textbf{x})}{\partial \textbf{x}} = \frac{\partial (\textbf{x}^T\textbf{a})}{\partial \textbf{x}} = \textbf{a} \\
\quad \textbf{a}^T\textbf{x} = \sum_i a_i x_i \implies \frac{\partial}{\partial x_k} = a_k \\ \quad \\
\text{2. } \frac{\partial (\textbf{x}^TA\textbf{x})}{\partial \textbf{x}} = (A + A^T)\textbf{x} \text{, A 对称时 } = 2A\textbf{x} \\
\quad \textbf{x}^TA\textbf{x} = \sum_i\sum_j a_{ij}x_ix_j \implies \frac{\partial}{\partial x_k} = \sum_j a_{kj}x_j + \sum_i a_{ik}x_i = (A\textbf{x})_k + (A^T\textbf{x})_k \\ \quad \\
\text{3. } \frac{\partial \|\textbf{y} - X\beta\|^2}{\partial \beta} = -2X^T(\textbf{y} - X\beta) \\
\quad \text{(multivariate linear regression 里推导过, 令其为 0 得到正规方程 } X^TX\beta = X^T\textbf{y} \text{)} \\ \quad \\
\text{4. Hessian 矩阵 (二阶导): } \frac{\partial^2 \|\textbf{y} - X\beta\|^2}{\partial \beta \partial \beta^T} = 2X^TX \text{ (半正定, 所以是最小值)} \\ \quad \\
\text{数值验证: 用中心差分 } \frac{\partial f}{\partial x_k} \approx \frac{f(\textbf{x} + h\textbf{e}_k) - f(\textbf{x} - h\textbf{e}_k)}{2h} \text{ 检查解析梯度, } \textbf{e}_k \text{ 是第 k 个单位向量}
$$

```python
def numerical_gradient(f, x, h=1e-6):
    grad = np.zeros_like(x, dtype=float)
    for k in range(len(x)):
        e_k = np.zeros_like(x, dtype=float)
        e_k[k] = 1.0
        grad[k] = (f(x + h * e_k) - f(x - h * e_k)) / (2 * h)
    return grad

def numerical_hessian(f, x, h=1e-4):
    n = len(x)
    H = np.zeros((n, n))
    for k in range(n):
        e_k = np.zeros(n)
        e_k[k] = 1.0
        H[:, k] = (numerical_gradient(f, x + h * e_k) - numerical_gradient(f, x - h * e_k)) / (2 * h)
    return H

rng = np.random.default_rng(1)
a_vec = rng.normal(size=3)
A_any = rng.normal(size=(3, 3))      # 不对称的矩阵
x0 = rng.normal(size=3)

print(f'1. d(a^T x)/dx:   解析 = {a_vec}, 数值 = {numerical_gradient(lambda x: a_vec @ x, x0)}')
print(f'2. d(x^T A x)/dx: 解析 = {(A_any + A_any.T) @ x0}, 数值 = {numerical_gradient(lambda x: x @ A_any @ x, x0)}')

X_design = np.column_stack([np.ones(10), rng.normal(size=(10, 2))])
y_vec = rng.normal(size=10)
beta0 = rng.normal(size=3)
loss = lambda b: np.sum((y_vec - X_design @ b) ** 2)
print(f'3. d||y - X b||^2/db: 解析 = {-2 * X_design.T @ (y_vec - X_design @ beta0)}, 数值 = {numerical_gradient(loss, beta0)}')
print(f'4. Hessian == 2 X^T X ? {np.allclose(numerical_hessian(loss, beta0), 2 * X_design.T @ X_design, atol=1e-4)}')
```

$$
\text{正交矩阵 (Orthogonal Matrix)} \\
Q^TQ = QQ^T = I \iff Q^{-1} = Q^T \text{ (求逆只需要转置)} \\
\text{Q 的每一列都是单位向量, 并且两两正交 (标准正交基)} \\ \quad \\
\text{性质} \\
\text{1. 保持长度: } \|Q\textbf{x}\|^2 = (Q\textbf{x})^T(Q\textbf{x}) = \textbf{x}^TQ^TQ\textbf{x} = \textbf{x}^T\textbf{x} = \|\textbf{x}\|^2 \\
\text{2. 保持内积 (所以也保持夹角): } (Q\textbf{x})^T(Q\textbf{y}) = \textbf{x}^TQ^TQ\textbf{y} = \textbf{x}^T\textbf{y} \\
\text{3. 行列式为 } \pm 1 \text{: } \det(Q^TQ) = \det(Q)^2 = \det(I) = 1 \text{ (+1 是旋转, -1 是反射)} \\ \quad \\
\text{例子: 二维旋转矩阵 } R(\theta) = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix} \\
R(\theta)^TR(\theta) = \begin{bmatrix} \cos^2\theta + \sin^2\theta & -\cos\theta\sin\theta + \sin\theta\cos\theta \\ -\sin\theta\cos\theta + \cos\theta\sin\theta & \sin^2\theta + \cos^2\theta \end{bmatrix} = I \\
\text{反射矩阵 (关于 x 轴): } F = \begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}, \ \det(F) = -1
$$

```python
import matplotlib.pyplot as plt

theta = np.pi / 6
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
F = np.array([[1, 0], [0, -1]])

print(f'R^T R = \n{np.round(R.T @ R, 10)}')
print(f'R^-1 == R^T ? {np.allclose(np.linalg.inv(R), R.T)}, det(R) = {np.linalg.det(R):.4f}, det(F) = {np.linalg.det(F):.4f}')

v1, v2 = np.array([3.0, 1.0]), np.array([1.0, 2.0])
print(f'|v1| = {np.linalg.norm(v1):.4f}, |R v1| = {np.linalg.norm(R @ v1):.4f}')
print(f'v1·v2 = {v1 @ v2:.4f}, (R v1)·(R v2) = {(R @ v1) @ (R @ v2):.4f}')

# 对称矩阵的特征向量矩阵就是正交矩阵
A_sym = np.array([[1, 2, 3, 4], [2, 5, 6, 7], [3, 6, 8, 9], [4, 7, 9, 10]], dtype=float)
eigvals_sym, Q_sym = np.linalg.eigh(A_sym)
print(f'对称矩阵特征向量组成的 Q: Q^T Q == I ? {np.allclose(Q_sym.T @ Q_sym, np.eye(4))}, Q Λ Q^T == A ? {np.allclose(Q_sym @ np.diag(eigvals_sym) @ Q_sym.T, A_sym)}')

shape = np.array([[0, 0], [2, 0], [2, 1], [0, 1], [0, 0]]).T    # 一个长方形
plt.figure(figsize=(4, 3))
plt.plot(*shape, 'b-', label='original')
plt.plot(*(R @ shape), 'r-', label='rotate 30 deg')
plt.plot(*(F @ shape), 'g--', label='reflect (x axis)')
plt.axis('equal')
plt.grid(True, alpha=0.3)
plt.legend(fontsize='small')
plt.tight_layout()
plt.show()
```

$$
\text{投影矩阵 (Projection Matrix) 与最小二乘的几何意义} \\
\text{最小二乘: } \hat{\beta} = (X^TX)^{-1}X^TY \implies \hat{Y} = X\hat{\beta} = \underbrace{X(X^TX)^{-1}X^T}_{H}Y = HY \\
\text{H 叫做帽子矩阵 (Hat Matrix), 因为它给 Y "戴上了帽子" 变成 } \hat{Y} \\ \quad \\
\text{性质} \\
\text{1. 对称: } H^T = H \\
\text{2. 幂等 (Idempotent): } H^2 = X(X^TX)^{-1}\underbrace{X^TX(X^TX)^{-1}}_{I}X^T = H \text{ (投影两次和投影一次一样)} \\
\text{3. 残差和 X 的每一列都正交: } X^T\textbf{e} = X^T(Y - \hat{Y}) = X^TY - X^TX\hat{\beta} = 0 \text{ (这就是正规方程)} \\
\text{4. H 的特征值只有 0 和 1 (} H\textbf{v} = \lambda\textbf{v} \implies H^2\textbf{v} = \lambda^2\textbf{v} = \lambda\textbf{v} \implies \lambda \in \{0, 1\} \text{), 所以 tr(H) = 秩 = X 的列数} \\ \quad \\
\text{几何意义: } \hat{Y} \text{ 是 Y 在 X 的列空间上的正交投影, 残差 } \textbf{e} \text{ 垂直于这个空间, 所以 } \|\textbf{e}\| \text{ 最短} \\
\text{勾股定理: } \|Y\|^2 = \|\hat{Y}\|^2 + \|\textbf{e}\|^2 \text{, 模型有截距项时, 把 Y 中心化之后就是 TSS = ESS + RSS}
$$

```python
# 用 linear_regression 里的身高-腿长数据
height = np.array([143, 145, 146, 147, 149, 150, 153, 154, 155, 156, 157, 158, 159, 160, 162, 164], dtype=float)
leg = np.array([88, 85, 88, 91, 92, 93, 93, 95, 96, 98, 97, 96, 98, 99, 100, 102], dtype=float)

X_reg = np.column_stack([np.ones_like(height), height])
H = X_reg @ np.linalg.inv(X_reg.T @ X_reg) @ X_reg.T
y_hat = H @ leg
e = leg - y_hat

print(f'H 对称? {np.allclose(H, H.T)}, H^2 == H? {np.allclose(H @ H, H)}, tr(H) = {np.trace(H):.4f}')
print(f'H 的特征值 (保留 4 位) = {np.unique(np.round(np.abs(np.linalg.eigvalsh(H)), 4))}')
print(f'X^T e = {X_reg.T @ e} (约等于 0)')

TSS = np.sum((leg - leg.mean()) ** 2)
ESS = np.sum((y_hat - leg.mean()) ** 2)
RSS = np.sum(e ** 2)
print(f'TSS = {TSS:.4f}, ESS + RSS = {ESS + RSS:.4f}')
```

$$
\text{QR 分解 (QR Decomposition)} \\
X = QR, \quad Q \text{ (n x k): 各列是标准正交的}, \quad R \text{ (k x k): 上三角矩阵} \\ \quad \\
\text{Gram-Schmidt 正交化: 依次把每一列减去它在前面所有方向上的投影, 再单位化} \\
\textbf{u}_j = \textbf{x}_j - \sum_{i<j} (\textbf{q}_i^T\textbf{x}_j)\textbf{q}_i, \quad \textbf{q}_j = \frac{\textbf{u}_j}{\|\textbf{u}_j\|} \\
r_{ij} = \textbf{q}_i^T\textbf{x}_j \ (i < j), \quad r_{jj} = \|\textbf{u}_j\| \\ \quad \\
\text{用 QR 解最小二乘: } X^TX\beta = X^TY \implies R^TQ^TQR\beta = R^TQ^TY \implies R\beta = Q^TY \quad (Q^TQ = I, R^T \text{ 可逆}) \\
\text{R 是上三角矩阵, 从最后一行往上回代就能解出 } \beta \text{, 不需要求 } (X^TX)^{-1} \\ \quad \\
\text{好处: 数值更稳定. 条件数 } \kappa(X^TX) = \kappa(X)^2 \text{, 直接求 } (X^TX)^{-1} \text{ 会把舍入误差放大很多}
$$

```python
def gram_schmidt(X):
    n, k = X.shape
    Q = np.zeros((n, k))
    R = np.zeros((k, k))
    for j in range(k):
        u = X[:, j].astype(float)
        for i in range(j):
            R[i, j] = Q[:, i] @ X[:, j]
            u = u - R[i, j] * Q[:, i]
        R[j, j] = np.linalg.norm(u)
        Q[:, j] = u / R[j, j]
    return Q, R

def back_substitution(R, b):
    k = len(b)
    x = np.zeros(k)
    for i in range(k - 1, -1, -1):
        x[i] = (b[i] - R[i, i + 1:] @ x[i + 1:]) / R[i, i]
    return x

Q_gs, R_gs = gram_schmidt(X_reg)
print(f'Q^T Q == I ? {np.allclose(Q_gs.T @ Q_gs, np.eye(2))}, QR == X ? {np.allclose(Q_gs @ R_gs, X_reg)}')
print(f'R = \n{R_gs}')

# numpy 的 QR 可能和手算的差一个正负号 (Q 的某列和 R 的对应行同时取反), 结果一样
Q_np, R_np = np.linalg.qr(X_reg)
signs = np.sign(np.diag(R_np))
print(f'和 np.linalg.qr 一致 (统一符号后)? {np.allclose(Q_np * signs, Q_gs) and np.allclose(R_np * signs[:, None], R_gs)}')

beta_qr = back_substitution(R_gs, Q_gs.T @ leg)
beta_lstsq = np.linalg.lstsq(X_reg, leg, rcond=None)[0]
print(f'QR 解出的 beta = {beta_qr}, lstsq = {beta_lstsq}')
print(f'cond(X) = {np.linalg.cond(X_reg):.2f}, cond(X^T X) = {np.linalg.cond(X_reg.T @ X_reg):.2f}, cond(X)^2 = {np.linalg.cond(X_reg) ** 2:.2f}')
```

$$
\text{奇异值分解 (Singular Value Decomposition, SVD)} \\
A = U\Sigma V^T \text{, 对任意 m x n 的矩阵都存在 (不要求是方阵, 也不要求可逆)} \\
U \text{ (m x m), V (n x n): 正交矩阵}, \quad \Sigma \text{: 对角线上是奇异值 } \sigma_1 \ge \sigma_2 \ge \cdots \ge 0 \\ \quad \\
\text{和特征值的关系: } A^TA = V\Sigma^TU^TU\Sigma V^T = V(\Sigma^T\Sigma)V^T \\
\text{所以 V 的列是 } A^TA \text{ 的特征向量, 奇异值 } \sigma_i = \sqrt{\lambda_i(A^TA)} \text{, 而 } \textbf{u}_i = \frac{A\textbf{v}_i}{\sigma_i} \\ \quad \\
\text{几何意义: 任何线性变换 = 旋转 } (V^T) \to \text{沿坐标轴拉伸 } (\Sigma) \to \text{再旋转 } (U) \\ \quad \\
\text{伪逆 (Moore-Penrose Pseudo-inverse): } A^+ = V\Sigma^+U^T \text{, } \Sigma^+ \text{ 把非零的奇异值取倒数, 零保持为零} \\
\text{最小二乘解: } \hat{\beta} = X^+Y \text{ (np.linalg.lstsq 和 np.linalg.pinv 内部用的就是 SVD, X 不满秩时也能求出最小范数解)} \\ \quad \\
\text{条件数 (Condition Number): } \kappa(A) = \frac{\sigma_{max}}{\sigma_{min}} \text{, 越大说明矩阵越接近奇异, 求解越不稳定} \\ \quad \\
\text{低秩近似: 只保留前 r 个奇异值 } A_r = \sum_{i=1}^r \sigma_i \textbf{u}_i\textbf{v}_i^T \text{, 它是所有秩为 r 的矩阵里最接近 A 的 (Eckart-Young 定理)} \\
\text{误差: } \|A - A_r\|_F = \sqrt{\sigma_{r+1}^2 + \cdots + \sigma_n^2}
$$

```python
rng = np.random.default_rng(2)
A_svd = rng.normal(size=(5, 3))

# 手动 SVD: 先求 A^T A 的特征值和特征向量
eigvals_ata, V = np.linalg.eigh(A_svd.T @ A_svd)
order = np.argsort(eigvals_ata)[::-1]               # 从大到小排列
eigvals_ata, V = eigvals_ata[order], V[:, order]
sigma = np.sqrt(eigvals_ata)
U = A_svd @ V / sigma                               # u_i = A v_i / sigma_i (精简版 U, 5 x 3)

U_np, sigma_np, Vt_np = np.linalg.svd(A_svd, full_matrices=False)
print(f'手动奇异值 = {sigma}')
print(f'numpy 奇异值 = {sigma_np}')
print(f'U Σ V^T == A ? {np.allclose(U @ np.diag(sigma) @ V.T, A_svd)}')
print(f'条件数 sigma_max / sigma_min = {sigma[0] / sigma[-1]:.4f}, np.linalg.cond = {np.linalg.cond(A_svd):.4f}')

# 伪逆求最小二乘
U_x, s_x, Vt_x = np.linalg.svd(X_reg, full_matrices=False)
X_pinv = Vt_x.T @ np.diag(1 / s_x) @ U_x.T
print(f'\n伪逆 beta = {X_pinv @ leg}, np.linalg.pinv = {np.linalg.pinv(X_reg) @ leg}')

# 共线的设计矩阵: 第 3 列 = 第 1 列 + 第 2 列, X^T X 不可逆, 但伪逆可以求解
X_col = np.column_stack([np.ones_like(height), height, 1 + height])
print(f'共线矩阵的奇异值 = {np.round(np.linalg.svd(X_col, compute_uv=False), 6)} (最后一个约等于 0)')
beta_min_norm = np.linalg.pinv(X_col) @ leg
print(f'最小范数解 = {beta_min_norm}, 预测值和满秩时一样? {np.allclose(X_col @ beta_min_norm, y_hat)}')

# 低秩近似
low_rank = np.outer(np.arange(1, 9), np.arange(1, 7)) + 0.5 * np.outer(np.sin(np.arange(8)), np.cos(np.arange(6)))
noisy = low_rank + rng.normal(0, 0.1, low_rank.shape)
U_n, s_n, Vt_n = np.linalg.svd(noisy, full_matrices=False)
for r in [1, 2, 3]:
    A_r = U_n[:, :r] @ np.diag(s_n[:r]) @ Vt_n[:r]
    print(f'r = {r}: ||A - A_r||_F = {np.linalg.norm(noisy - A_r):.4f}, sqrt(剩余奇异值平方和) = {np.sqrt(np.sum(s_n[r:] ** 2)):.4f}')

plt.figure(figsize=(4, 3))
plt.plot(range(1, len(s_n) + 1), s_n, 'bo-', label='singular values')
plt.yscale('log')
plt.xlabel('index')
plt.title('Rank-2 signal + noise')
plt.grid(True, which='both', alpha=0.3)
plt.legend(fontsize='small')
plt.tight_layout()
plt.show()
```
