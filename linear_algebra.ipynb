$$
\text{矩阵的秩 (Rank) r(A): 指的是有效信息的行. 经过初等行变换之后, 剩下的个数就为这个矩阵的秩} \\
$$


$$
矩阵的零度 (Nullity): 指的是矩阵的解空间的维数 \\
和秩的关系：\\
\text{Rank(A) + Nullity(A) = n} \\
\text{其中，n 是矩阵的列数} \\
$$


$$
\text{矩阵的迹 (Trace) tr(A): 指方阵对角线上所有元素之和} \\
tr(A) = \sum_{i=1}^n A_{ii} \\
当A矩阵的秩为1时, 零度为n-1, 也就是说除一个元素外, 其他元素都为0, 那么 \\
tr(A) = \lambda + 0 + 0 + ... + 0 \\
tr(A) = \lambda \\
\text{其中} \lambda \text{为A的一个特征值} \\
$$


$$
\text{矩阵的行列式 (Determinant): 指的是一个方阵的特征值的乘积, 可以用于判断矩阵是否可逆, 以及缩放比例} \\
det(A) = GaussianElimination(A) -> \text{最终得到的上三角矩阵的对角线元素的乘积} \\
det(A) != 0 \text{ 时, 矩阵可逆, 且缩放比例为 det(A)} \\
det(A) = 0 \text{ 时, 矩阵变换后, 会导致降维, 继而不可逆} \\
$$


$$
\text{矩阵的逆 (Inverse): 指的是使得是n阶方阵A乘以其逆矩阵得到单位矩阵I} \\
\text{Gauss-Jordan 消元法 (初等变换): 是一种用于求解线性方程组的方法, 同时也可以用于求矩阵的逆} \\
[A|I] = [I|A^{-1}] \\
例子: \\
[A|I] = \begin{bmatrix}
1 & 2 & 3 & 1 & 0 & 0 \\
0 & 1 & 4 & 0 & 1 & 0 \\
5 & 6 & 0 & 0 & 0 & 1 \\
\end{bmatrix} \\
[I|A^{-1}] = \begin{bmatrix}
1 & 0 & 0 & -24 & 18 & 5 \\
0 & 1 & 0 & 20 & -15 & -4 \\
0 & 0 & 1 & -5 & 4 & 1 \\
\end{bmatrix} \\
$$


$$
矩阵乘法的数学通式 \\
C_{ij} = \sum_{k=1}^{n} A_{ik}B_{kj} \\
C = \begin{bmatrix}
    \sum_{k=1}^{n} A_{1k}B_{k1} & \sum_{k=1}^{n} A_{1k}B_{k2} & \cdots & \sum_{k=1}^{n} A_{1k}B_{kn} \\
    \sum_{k=1}^{n} A_{2k}B_{k1} & \sum_{k=1}^{n} A_{2k}B_{k2} & \cdots & \sum_{k=1}^{n} A_{2k}B_{kn} \\
    \vdots & \vdots & \ddots & \vdots \\
    \sum_{k=1}^{n} A_{nk}B_{k1} & \sum_{k=1}^{n} A_{nk}B_{k2} & \cdots & \sum_{k=1}^{n} A_{nk}B_{kn} \\
\end{bmatrix} \\
如果B是列向量 \\
C = \begin{bmatrix}
    \sum_{k=1}^{n} A_{1k}B_{k1} \\
    \sum_{k=1}^{n} A_{2k}B_{k1} \\
    \vdots \\
    \sum_{k=1}^{n} A_{nk}B_{k1} \\
\end{bmatrix} \\
$$

$$
\text{矩阵的转置 (Transpose): 指的是将矩阵的行和列互换得到的新矩阵} \\
A_{ij}^T = A_{ji} \\
\quad \\
(A + Bx)^T = A^T + (Bx)^T = A^T + x^TB^T \\
(A + BCx)^T = A^T + (BCx)^T = A^T + x^TC^TB^T \\
$$


$$
\text{矩阵的特征值 (Eigenvalue): 指的是在矩阵变换下, 向量长度发生改变的比例} \\
\textbf{Ax} = \lambda I \textbf{x} \\
(\textbf{A} - \lambda I)\textbf{x} = 0 \\
\text{其中, A 是矩阵, x 是特征向量, I 是单位矩阵} \\
\text{对于Mx=0, 有两种方法, 一是M矩阵可逆, 那x只有唯一解0, 要使x非零解, 那M矩阵必须是不可逆也就是奇异矩阵} \\
\text{不可逆的充分必要条件即是行列式为0} \\
\det(\textbf{A} - \lambda I) = 0 \\
\lambda\text{ 是特征值, 要找的就是这个缩放后的}\lambda \\

\text{矩阵的特征向量 (Eigenvector): 指的是在矩阵变换下, 方向保持不变的向量} \\
\text{将求得的特征值代入到 }(\textbf{A} - \lambda I)\textbf{x} = 0 \text{中, 求得x即可得到对应的特征向量} \\
$$


$$
\text{对称矩阵 Symmetric Matrix} \\
\quad \\
A = \begin{bmatrix}
1 & 2 & 3 & 4 \\
2 & 5 & 6 & 7 \\
3 & 6 & 8 & 9 \\
4 & 7 & 9 & 10 \\
\end{bmatrix} = A^T \\
\quad \\
\text{对称矩阵的性质} \\
\text{1. 对称矩阵的主对角线元素都是非负数} \\
\text{2. 对称矩阵的元素满足} a_{ij} = a_{ji} \\
\text{3. 对称矩阵的特征值都是非负数} \\
\text{4. 对称矩阵的特征向量是正交的} \\
\quad \\
$$

$$
\text{对角矩阵 Diagonal Matrix} \\
\quad \\
A = \begin{bmatrix}
a_{11} & 0 & \cdots & 0 \\
0 & a_{22} & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & a_{nn}
\end{bmatrix} = A^T \\
\text{一定是对称矩阵}
\quad \\
$$

$$
\text{矩阵的对角化 (Diagonalization): 指的是将一个矩阵表示为其特征向量的线性组合, 其中每个特征向量都是一个基向量} \\
A = PDP^{-1} \\
\text{其中, P 是特征向量矩阵, D 是特征值矩阵} \\
$$

$$
\text{幂零矩阵 (Nilpotent Matrix): 指一个矩阵A, 满足} A^n = 0 \text{, 其中n为一个整数, 且n大于等于1} \\
\text{严格上(下)三角矩阵：主对角线及其以下(以上)元素全为0的矩阵} \\
\begin{bmatrix}
    0 & a_{12} & \cdots & a_{1n} \\
    0 & 0 & \cdots & a_{2n} \\
    \vdots & \vdots & \ddots & \vdots \\
    0 & 0 & \cdots & 0 \\
\end{bmatrix}^2 = 
\begin{bmatrix}
    0 & 0 & \cdots & 0 \\
    0 & 0 & \cdots & 0 \\
    \vdots & \vdots & \ddots & \vdots \\
    0 & 0 & \cdots & 0 \\
\end{bmatrix}
$$
