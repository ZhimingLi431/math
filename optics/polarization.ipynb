<a id="jones-vector"></a>

$$
\text{偏振 (Polarization): 光是横波, 电场在垂直于传播方向 (z) 的平面内振动} \\
E(z, t) = \text{Re}\left\{\begin{bmatrix} \tilde{E}_x \\ \tilde{E}_y \end{bmatrix} e^{i(kz - \omega t)}\right\} \\
\text{琼斯向量 (Jones Vector): 只记录两个分量的复振幅 } J = \begin{bmatrix} \tilde{E}_x \\ \tilde{E}_y \end{bmatrix} \text{, 强度 } I = |\tilde{E}_x|^2 + |\tilde{E}_y|^2 = J^H J \\ \quad \\
\text{常见偏振态:} \\
\text{水平线偏振 } \begin{bmatrix} 1 \\ 0 \end{bmatrix}, \quad \text{竖直 } \begin{bmatrix} 0 \\ 1 \end{bmatrix}, \quad \text{角度 } \theta \text{ 的线偏振 } \begin{bmatrix} \cos\theta \\ \sin\theta \end{bmatrix} \ (\text{两个分量同相}) \\
\text{圆偏振 (Circular) } \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ \pm i \end{bmatrix} \text{: 振幅相等, 相位差 } \pm\frac{\pi}{2} \\
\text{验证: } E_x = \text{Re}(e^{-i\omega t}) = \cos\omega t, \ E_y = \text{Re}(i e^{-i\omega t}) = \text{Re}(e^{i(\pi/2 - \omega t)}) = \sin\omega t \text{, 电场矢量画圆} \\ \quad \\
\text{光学元件 = 2x2 复矩阵 (琼斯矩阵 Jones Matrix), 光依次通过多个元件 = 矩阵从右往左相乘 (和 ABCD 矩阵一样)}
$$

数学基础: [欧拉公式](../math_basics/calculus.ipynb#taylor) · [矩阵乘法](../math_basics/linear_algebra.ipynb#matrix-multiplication) · 应用: [ABCD 矩阵](thin_lens_and_abcd_matrix.ipynb#abcd)

<a id="polarizer-malus"></a>

$$
\text{偏振片 (Polarizer): 只让沿透光轴 (单位向量 u) 的分量通过} \\
J_{\text{out}} = u \, (u^T J) = (u u^T) J \implies P_\theta = u u^T = \begin{bmatrix} \cos^2\theta & \cos\theta\sin\theta \\ \cos\theta\sin\theta & \sin^2\theta \end{bmatrix}, \quad u = \begin{bmatrix} \cos\theta \\ \sin\theta \end{bmatrix} \\
\text{这就是投影矩阵 (Projection Matrix): 对称 } P^T = P \text{, 幂等 } P^2 = u(u^T u)u^T = P \text{ (过两次相同的偏振片, 第二次没有变化)} \\ \quad \\
\text{马吕斯定律 (Malus's Law): 线偏振光 (方向 } \theta_0\text{) 通过透光轴为 } \theta \text{ 的偏振片} \\
u^T J = \cos\theta\cos\theta_0 + \sin\theta\sin\theta_0 = \cos(\theta - \theta_0) \\
I = |u^T J|^2 = I_0\cos^2(\theta - \theta_0) \\ \quad \\
\text{自然光 (Unpolarized Light): 偏振方向随机, 对 } \theta_0 \text{ 取平均: } \langle\cos^2\rangle = \frac{1}{2} \implies I = \frac{I_0}{2} \\ \quad \\
\text{旋转矩阵 (Rotation Matrix): } R(\theta) = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix} \text{, 正交矩阵 } R^T R = I, \ R^{-1} = R(-\theta) \\
\text{任意角度的元件 = 先把坐标系转到元件的轴上, 用轴对齐的矩阵, 再转回来: } M_\theta = R(\theta) M_0 R(-\theta) \\
\text{例: } P_\theta = R(\theta)\begin{bmatrix} 1 & 0 \\ 0 & 0 \end{bmatrix}R(-\theta) \text{ (相似变换, 和对角化 } A = P D P^{-1} \text{ 同一个形式)}
$$

数学基础: [投影矩阵](../math_basics/linear_algebra.ipynb#projection) · [正交矩阵与旋转](../math_basics/linear_algebra.ipynb#orthogonal) · [对角化](../math_basics/linear_algebra.ipynb#diagonalization) · [期望](../math_basics/statistics.ipynb#expectation)

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(linewidth=np.inf)
np.set_printoptions(suppress=True)

def plot_nicely():
    for ax in plt.gcf().get_axes():
        ax.grid(True, which="both", alpha=0.3)
        if ax.get_legend_handles_labels()[0]:
            ax.legend(fontsize='small')
    plt.tight_layout()

def rot(theta):
    return np.array([[np.cos(theta), -np.sin(theta)], [np.sin(theta), np.cos(theta)]])

def polarizer(theta):
    u = np.array([np.cos(theta), np.sin(theta)])
    return np.outer(u, u)

def linear(theta):
    return np.array([np.cos(theta), np.sin(theta)], dtype=complex)

def intensity(J):
    return float(np.real(np.vdot(J, J)))   # J^H J

th = np.deg2rad(30)
P30 = polarizer(th)
print(f'P_30 = \n{P30}')
print(f'对称? {np.allclose(P30, P30.T)}, 幂等 P^2 = P? {np.allclose(P30 @ P30, P30)}, 特征值 = {np.round(np.linalg.eigvalsh(P30), 10) + 0.0}')
print(f'u u^T == R(theta) P_0 R(-theta) ? {np.allclose(P30, rot(th) @ polarizer(0) @ rot(-th))}')
print(f'R 是正交矩阵? {np.allclose(rot(th).T @ rot(th), np.eye(2))}, det = {np.linalg.det(rot(th)):.6f}')

# 马吕斯定律
angles = np.linspace(0, np.pi, 181)
I_out = np.array([intensity(polarizer(a) @ linear(0)) for a in angles])
print(f'马吕斯定律 I = cos^2(theta) ? {np.allclose(I_out, np.cos(angles)**2)}')

# 自然光: 随机偏振方向取平均
rng = np.random.default_rng(0)
random_dirs = rng.uniform(0, 2 * np.pi, 20000)
I_unpolarized = np.mean([intensity(P30 @ linear(t)) for t in random_dirs[:20000]])   # 每个方向都用琼斯矩阵算一遍
print(f'自然光通过偏振片: 平均强度 = {I_unpolarized:.4f} (理论 0.5)')

plt.figure(figsize=(4, 3))
plt.plot(np.degrees(angles), I_out, label='Jones matrix')
plt.plot(np.degrees(angles[::10]), np.cos(angles[::10])**2, 'o', markersize=3, label='cos^2 theta')
plt.xlabel('polarizer angle (deg)')
plt.ylabel('I / I0')
plt.title("Malus's law")
plot_nicely()
plt.show()
```

<a id="wave-plates"></a>

$$
\text{波片 (Wave Plate / Retarder): 双折射晶体, 快轴 (fast axis) 和慢轴方向的折射率不同} \\
\text{厚度 L 的晶体, 两个分量的相位差 } \Gamma = \frac{2\pi}{\lambda}(n_{\text{slow}} - n_{\text{fast}}) L \\
\text{快轴沿 x: } W(\Gamma) = \begin{bmatrix} 1 & 0 \\ 0 & e^{i\Gamma} \end{bmatrix} \text{ (省略两个分量共同的相位)} \\
\text{四分之一波片 (Quarter-Wave Plate, QWP): } \Gamma = \frac{\pi}{2}, \ W = \text{diag}(1, i) \\
\text{二分之一波片 (Half-Wave Plate, HWP): } \Gamma = \pi, \ W = \text{diag}(1, -1) \\ \quad \\
\text{1. 快轴在 } \theta \text{ 的半波片: } H_\theta = R(\theta)\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}R(-\theta) \\
\text{用二倍角公式 } \cos^2\theta - \sin^2\theta = \cos 2\theta, \ 2\sin\theta\cos\theta = \sin 2\theta \text{:} \\
H_\theta = \begin{bmatrix} \cos 2\theta & \sin 2\theta \\ \sin 2\theta & -\cos 2\theta \end{bmatrix} \quad (\text{关于 } \theta \text{ 方向的反射矩阵}) \\
H_\theta \begin{bmatrix} \cos\theta_0 \\ \sin\theta_0 \end{bmatrix} = \begin{bmatrix} \cos(2\theta - \theta_0) \\ \sin(2\theta - \theta_0) \end{bmatrix} \text{: 线偏振方向从 } \theta_0 \text{ 变成 } 2\theta - \theta_0 \\
\text{(水平偏振光通过 } \theta \text{ 的半波片, 偏振方向转了 } 2\theta \text{, 不损失能量, 用来旋转偏振方向)} \\ \quad \\
\text{2. 快轴在 45° 的四分之一波片 + 水平线偏振光:} \\
Q_{45} = R(45^\circ)\begin{bmatrix} 1 & 0 \\ 0 & i \end{bmatrix}R(-45^\circ) = \frac{1}{2}\begin{bmatrix} 1 + i & 1 - i \\ 1 - i & 1 + i \end{bmatrix} \\
Q_{45}\begin{bmatrix} 1 \\ 0 \end{bmatrix} = \frac{1}{2}\begin{bmatrix} 1 + i \\ 1 - i \end{bmatrix} = \frac{1 + i}{2}\begin{bmatrix} 1 \\ -i \end{bmatrix} \quad (\text{因为 } \frac{1 - i}{1 + i} = \frac{(1 - i)^2}{2} = -i) \\
\text{两个分量振幅相等, 相位差 } \frac{\pi}{2} \text{: 线偏振 → 圆偏振 (3D 电影眼镜, 光隔离器)} \\ \quad \\
\text{波片是幺正矩阵 (Unitary, } W^H W = I\text{): 保持强度 } J^H W^H W J = J^H J \text{ (正交矩阵在复数上的推广)}
$$

```python
def wave_plate(gamma, theta=0.0):
    return rot(theta) @ np.diag([1, np.exp(1j * gamma)]) @ rot(-theta)

# 1. 半波片 = 反射矩阵, 偏振方向 theta0 -> 2 theta - theta0
for th_hwp in np.deg2rad([10, 22.5, 45]):
    H = wave_plate(np.pi, th_hwp)
    reflection = np.array([[np.cos(2 * th_hwp), np.sin(2 * th_hwp)], [np.sin(2 * th_hwp), -np.cos(2 * th_hwp)]])
    out = H @ linear(np.deg2rad(5))
    angle_out = np.degrees(np.arctan2(out[1].real, out[0].real))
    print(f'HWP {np.degrees(th_hwp):>4.1f}°: == 反射矩阵? {np.allclose(H, reflection)}, 5° 线偏振 -> {angle_out:.4f}° (公式 2 theta - 5 = {2 * np.degrees(th_hwp) - 5:.4f}°), 强度 = {intensity(out):.6f}')

# 2. 45° 四分之一波片: 水平 -> 圆偏振
Q45 = wave_plate(np.pi / 2, np.pi / 4)
print(f'Q45 = \n{np.round(Q45, 6)}\n== (1/2)[[1+i, 1-i], [1-i, 1+i]] ? {np.allclose(Q45, 0.5 * np.array([[1 + 1j, 1 - 1j], [1 - 1j, 1 + 1j]]))}')
J_circ = Q45 @ linear(0)
print(f'输出 = {np.round(J_circ, 6)}, |Ex| = {abs(J_circ[0]):.6f}, |Ey| = {abs(J_circ[1]):.6f}, 相位差 = {np.degrees(np.angle(J_circ[1] / J_circ[0])):.2f}°')
print(f'幺正? Q^H Q = I : {np.allclose(Q45.conj().T @ Q45, np.eye(2))}')

# 圆偏振通过任意角度的偏振片: 强度都是一半
print(f'圆偏振通过 0°, 30°, 77° 偏振片: {[round(intensity(polarizer(np.deg2rad(a)) @ J_circ), 6) for a in [0, 30, 77]]}')
# 两个 45° 四分之一波片 = 一个半波片 (相位延迟相加)
print(f'Q45 Q45 == H45 ? {np.allclose(Q45 @ Q45, wave_plate(np.pi, np.pi / 4))}')

# 电场矢量随时间的轨迹 Re[J e^{-i w t}]
wt = np.linspace(0, 2 * np.pi, 200)
fig, axs = plt.subplots(1, 3, figsize=(8, 3))
for ax, (name, J) in zip(axs, [('linear 30 deg', linear(np.deg2rad(30))), ('circular (QWP)', J_circ),
                               ('elliptical', wave_plate(np.pi / 4, np.pi / 4) @ linear(0))]):
    field = np.real(np.outer(J, np.exp(-1j * wt)))
    ax.plot(field[0], field[1])
    ax.set_xlim(-1.1, 1.1); ax.set_ylim(-1.1, 1.1)
    ax.set_aspect('equal')
    ax.set_title(name)
    ax.set_xlabel('Ex'); ax.set_ylabel('Ey')
plt.tight_layout()
plt.show()
```

<a id="three-polarizers"></a>

$$
\text{三偏振片佯谬 (Three-Polarizer Paradox)} \\
\text{水平偏振光通过竖直偏振片: } I = \cos^2 90^\circ = 0 \text{ (正交, 完全挡住)} \\
\text{在两者之间插入一个 45° 偏振片 (多加一个 "挡光" 的元件), 反而有光通过:} \\
I = \cos^2 45^\circ \cdot \cos^2 45^\circ = \frac{1}{4} \\
\text{原因: 投影矩阵不可交换 } P_{90} P_{45} P_0 \neq P_{90} P_0 P_{45} \text{, 而 } P_{90} P_0 = 0 \\ \quad \\
\text{推广: 用 N 个偏振片把 90° 等分, 每两个相邻偏振片之间夹角 } \frac{\pi}{2N} \\
I_N = \left[\cos^2\frac{\pi}{2N}\right]^N \\
\text{取对数, 用泰勒展开 } \ln\cos x \approx \ln\left(1 - \frac{x^2}{2}\right) \approx -\frac{x^2}{2} \text{:} \\
\ln I_N = 2N\ln\cos\frac{\pi}{2N} \approx 2N \cdot \left(-\frac{\pi^2}{8N^2}\right) = -\frac{\pi^2}{4N} \to 0 \implies I_N \to 1 \\
\text{偏振片足够多时, 可以几乎无损地把偏振方向转 90° (和量子芝诺效应 Quantum Zeno Effect 是同一个数学结构)}
$$

```python
J0 = linear(0)
I_crossed = intensity(polarizer(np.pi / 2) @ polarizer(0) @ J0)
I_three = intensity(polarizer(np.pi / 2) @ polarizer(np.pi / 4) @ polarizer(0) @ J0)
print(f'0° + 90°: I = {I_crossed:.6f}')
print(f'0° + 45° + 90°: I = {I_three:.6f} (理论 1/4)')
print(f'P90 P0 = \n{polarizer(np.pi / 2) @ polarizer(0) + 0.0}\nP45 P0 == P0 P45 ? {np.allclose(polarizer(np.pi / 4) @ polarizer(0), polarizer(0) @ polarizer(np.pi / 4))} (不可交换)')

Ns = np.arange(1, 101)
I_N = []
for N in Ns:
    J = J0.copy()
    for j in range(1, N + 1):
        J = polarizer(j * np.pi / (2 * N)) @ J
    I_N.append(intensity(J))
I_N = np.array(I_N)
print(f'N 个偏振片的矩阵乘积 == cos^(2N)(pi / 2N) ? {np.allclose(I_N, np.cos(np.pi / (2 * Ns))**(2 * Ns))}')
for N in [2, 10, 100]:
    print(f'N = {N:>3}: I = {I_N[N - 1]:.6f}, 近似 exp(-pi^2 / 4N) = {np.exp(-np.pi**2 / (4 * N)):.6f}')

plt.figure(figsize=(4, 3))
plt.plot(Ns, I_N, 'o', markersize=3, label='product of N polarizers')
plt.plot(Ns, np.exp(-np.pi**2 / (4 * Ns)), label='exp(-pi^2 / 4N)')
plt.xscale('log')
plt.xlabel('N')
plt.ylabel('transmitted I / I0')
plot_nicely()
plt.show()
```
