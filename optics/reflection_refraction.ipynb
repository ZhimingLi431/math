<a id="fermat-snell"></a>

$$
\text{费马原理 (Fermat's Principle): 光在两点之间走 "用时最短" (严格说是用时取极值) 的路径} \\
\text{折射率 (Refractive Index) } n = \frac{c}{v} \text{: 光在介质中的速度 } v = \frac{c}{n} \text{, 光程 (Optical Path Length) = } n \times \text{几何长度} \\ \quad \\
\text{推导斯涅尔定律 (Snell's Law):} \\
\text{A = (0, } h_1) \text{ 在介质 } n_1 \text{ 中, B = (d, } -h_2) \text{ 在介质 } n_2 \text{ 中, 界面是 y = 0, 光在 (x, 0) 处穿过界面} \\
t(x) = \frac{n_1\sqrt{x^2 + h_1^2}}{c} + \frac{n_2\sqrt{(d - x)^2 + h_2^2}}{c} \\
\frac{dt}{dx} = \frac{1}{c}\left[\frac{n_1 x}{\sqrt{x^2 + h_1^2}} - \frac{n_2 (d - x)}{\sqrt{(d - x)^2 + h_2^2}}\right] = 0 \quad (\text{链式法则}) \\
\text{几何上 } \frac{x}{\sqrt{x^2 + h_1^2}} = \sin\theta_1, \ \frac{d - x}{\sqrt{(d - x)^2 + h_2^2}} = \sin\theta_2 \quad (\theta \text{ 是和法线的夹角}) \\
n_1\sin\theta_1 = n_2\sin\theta_2 \\ \quad \\
\text{是最小值: } \frac{d^2 t}{dx^2} = \frac{1}{c}\left[\frac{n_1 h_1^2}{(x^2 + h_1^2)^{3/2}} + \frac{n_2 h_2^2}{((d - x)^2 + h_2^2)^{3/2}}\right] > 0 \\ \quad \\
\text{反射定律 (Law of Reflection): 同一种介质, 把 B 关于镜面对称到 B' , 路径长度不变} \\
\text{A 到 B' 的最短路径是直线 } \implies \theta_i = \theta_r \text{ (入射角 = 反射角)}
$$

数学基础: [链式法则](../math_basics/calculus.ipynb#derivative-rules) · [极值的一阶 / 二阶条件](../math_basics/calculus.ipynb#extremum)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar, brentq

np.set_printoptions(linewidth=np.inf)
np.set_printoptions(suppress=True)

def plot_nicely():
    for ax in plt.gcf().get_axes():
        ax.grid(True, which="both", alpha=0.3)
        if ax.get_legend_handles_labels()[0]:
            ax.legend(fontsize='small')
    plt.tight_layout()

c = 1.0
n1, n2 = 1.0, 1.5   # 空气 -> 玻璃
h1, h2, d = 1.0, 1.0, 2.0

travel_time = lambda x: n1 * np.sqrt(x**2 + h1**2) / c + n2 * np.sqrt((d - x)**2 + h2**2) / c
res = minimize_scalar(travel_time, bounds=(0, d), method='bounded', options={'xatol': 1e-12})
x_star = res.x
sin1 = x_star / np.sqrt(x_star**2 + h1**2)
sin2 = (d - x_star) / np.sqrt((d - x_star)**2 + h2**2)
print(f'用时最短的穿越点 x = {x_star:.8f}')
print(f'n1 sin(theta1) = {n1 * sin1:.8f}, n2 sin(theta2) = {n2 * sin2:.8f}')
print(f'入射角 = {np.degrees(np.arcsin(sin1)):.4f}°, 折射角 = {np.degrees(np.arcsin(sin2)):.4f}°')

# 二阶导 > 0
h = 1e-4
print(f't\'\'(x*) 数值 = {(travel_time(x_star + h) - 2 * travel_time(x_star) + travel_time(x_star - h)) / h**2:.6f}, '
      f'公式 = {(n1 * h1**2 / (x_star**2 + h1**2)**1.5 + n2 * h2**2 / ((d - x_star)**2 + h2**2)**1.5) / c:.6f}')

# 直线 (几何最短) 不是用时最短
print(f'走直线 (x = d/2 = {d / 2}) 用时 = {travel_time(d / 2):.6f}, 最优路径用时 = {res.fun:.6f}')

xs = np.linspace(0, d, 200)
fig, axs = plt.subplots(1, 2, figsize=(8, 3))
axs[0].plot(xs, travel_time(xs), label='t(x)')
axs[0].axvline(x_star, color='r', linestyle='--', label='minimum')
axs[0].set_xlabel('crossing point x')
axs[0].set_ylabel('travel time')
axs[1].axhspan(-1.3, 0, color='lightblue', alpha=0.4, label=f'n2 = {n2}')
axs[1].plot([0, x_star, d], [h1, 0, -h2], 'r-o', label='fastest path')
axs[1].plot([0, d], [h1, -h2], 'k--', label='straight line')
axs[1].set_aspect('equal')
plot_nicely()
plt.show()
```

<a id="total-internal-reflection"></a>

$$
\text{全反射 (Total Internal Reflection): 从光密介质射向光疏介质 } (n_1 > n_2) \\
\sin\theta_2 = \frac{n_1}{n_2}\sin\theta_1 \text{, 当 } \sin\theta_1 > \frac{n_2}{n_1} \text{ 时 } \sin\theta_2 > 1 \text{, 没有实数的折射角, 光全部被反射} \\
\text{临界角 (Critical Angle): } \theta_c = \arcsin\frac{n_2}{n_1} \quad (\text{玻璃→空气: } \arcsin\frac{1}{1.5} \approx 41.8^\circ) \\
\text{应用: 光纤 (Optical Fiber), 全反射棱镜}
$$

<a id="fresnel"></a>

$$
\text{菲涅尔公式 (Fresnel Equations): 反射和透射了多少能量} \\ \quad \\
\text{推导思路 (s 偏振, 电场垂直于入射面): 在界面上电场和磁场的切向分量连续 (由麦克斯韦方程组得到)} \\
\text{电场切向: } E_i + E_r = E_t \\
\text{磁场切向: 介质中 } H \propto n E \text{, 方向用右手定则, 投影到界面上带 } \cos\theta \text{: } n_1(E_i - E_r)\cos\theta_i = n_2 E_t \cos\theta_t \\
\text{把第一个式子代入第二个: } n_1(E_i - E_r)\cos\theta_i = n_2(E_i + E_r)\cos\theta_t \\
r_s = \frac{E_r}{E_i} = \frac{n_1\cos\theta_i - n_2\cos\theta_t}{n_1\cos\theta_i + n_2\cos\theta_t}, \quad t_s = \frac{E_t}{E_i} = 1 + r_s = \frac{2 n_1\cos\theta_i}{n_1\cos\theta_i + n_2\cos\theta_t} \\
\text{p 偏振 (电场在入射面内) 同理: } r_p = \frac{n_2\cos\theta_i - n_1\cos\theta_t}{n_2\cos\theta_i + n_1\cos\theta_t}, \quad t_p = \frac{2 n_1\cos\theta_i}{n_2\cos\theta_i + n_1\cos\theta_t} \\ \quad \\
\text{反射率 (Reflectance) } R = |r|^2 \text{, 透射率 (Transmittance) } T = \frac{n_2\cos\theta_t}{n_1\cos\theta_i}|t|^2 \\
\text{(T 多了一个因子: 能流正比于 } n|E|^2 \text{, 并且光束截面积按 } \cos\theta \text{ 缩放), 能量守恒 } R + T = 1 \\
\text{垂直入射 } \theta_i = 0 \text{: } R = \left(\frac{n_1 - n_2}{n_1 + n_2}\right)^2 \quad (\text{空气-玻璃约 4\%}) \\
\text{全反射时 } \cos\theta_t = \sqrt{1 - \sin^2\theta_t} \text{ 变成纯虚数, r 的分子分母互为共轭, } |r| = 1
$$

<a id="brewster"></a>

$$
\text{布儒斯特角 (Brewster's Angle): p 偏振的反射率为 0} \\
r_p = 0 \iff n_2\cos\theta_i = n_1\cos\theta_t \\
\text{两边平方, 用斯涅尔定律 } \sin\theta_t = \frac{n_1}{n_2}\sin\theta_i \text{, 记 } s = \sin^2\theta_i \text{:} \\
n_2^2(1 - s) = n_1^2\left(1 - \frac{n_1^2}{n_2^2}s\right) \implies n_2^2 - n_1^2 = s\left(n_2^2 - \frac{n_1^4}{n_2^2}\right) = s\frac{(n_2^2 - n_1^2)(n_2^2 + n_1^2)}{n_2^2} \\
s = \frac{n_2^2}{n_1^2 + n_2^2}, \quad 1 - s = \frac{n_1^2}{n_1^2 + n_2^2} \implies \tan^2\theta_B = \frac{s}{1 - s} = \frac{n_2^2}{n_1^2} \\
\tan\theta_B = \frac{n_2}{n_1} \quad (\text{空气→玻璃约 } 56.3^\circ) \\ \quad \\
\text{此时反射光和折射光垂直: } \sin\theta_t = \frac{n_1}{n_2}\sin\theta_B = \frac{n_1}{n_2} \cdot \frac{n_2}{\sqrt{n_1^2 + n_2^2}} = \cos\theta_B \implies \theta_B + \theta_t = 90^\circ \\
\text{应用: 以布儒斯特角反射的光只剩 s 偏振 (偏振太阳镜, 拍照消除水面反光)}
$$

```python
def fresnel(theta_i, n1, n2):
    sin_t = n1 / n2 * np.sin(theta_i)
    cos_i = np.cos(theta_i)
    cos_t = np.sqrt(1 - sin_t.astype(complex)**2)   # 全反射时为纯虚数
    rs = (n1 * cos_i - n2 * cos_t) / (n1 * cos_i + n2 * cos_t)
    rp = (n2 * cos_i - n1 * cos_t) / (n2 * cos_i + n1 * cos_t)
    ts = 2 * n1 * cos_i / (n1 * cos_i + n2 * cos_t)
    tp = 2 * n1 * cos_i / (n2 * cos_i + n1 * cos_t)
    factor = np.real(n2 * cos_t) / (n1 * cos_i)
    return np.abs(rs)**2, np.abs(rp)**2, factor * np.abs(ts)**2, factor * np.abs(tp)**2

theta = np.linspace(0, np.pi / 2 - 1e-6, 2000)

# 空气 -> 玻璃
Rs, Rp, Ts, Tp = fresnel(theta, 1.0, 1.5)
print(f'能量守恒 R + T = 1 ? s: {np.allclose(Rs + Ts, 1)}, p: {np.allclose(Rp + Tp, 1)}')
print(f'垂直入射 R = {Rs[0]:.6f}, 公式 ((n1 - n2) / (n1 + n2))^2 = {((1 - 1.5) / (1 + 1.5))**2:.6f}')
theta_B_num = brentq(lambda th: (1.5 * np.cos(th) - np.cos(np.arcsin(np.sin(th) / 1.5))), 0.1, 1.5)   # r_p 的分子 = 0
theta_B = np.arctan(1.5 / 1.0)
theta_t_B = np.arcsin(np.sin(theta_B) / 1.5)
print(f'布儒斯特角: 求根 = {np.degrees(theta_B_num):.6f}°, arctan(n2 / n1) = {np.degrees(theta_B):.6f}°, R_p = {fresnel(np.array([theta_B]), 1.0, 1.5)[1][0]:.2e}')
print(f'theta_B + theta_t = {np.degrees(theta_B + theta_t_B):.6f}°')

# 玻璃 -> 空气: 全反射
Rs2, Rp2, _, _ = fresnel(theta, 1.5, 1.0)
theta_c = np.arcsin(1.0 / 1.5)
beyond = theta > theta_c + 1e-9
print(f'临界角 = {np.degrees(theta_c):.4f}°, 超过临界角后 R_s 和 R_p 都等于 1 ? {np.allclose(Rs2[beyond], 1) and np.allclose(Rp2[beyond], 1)}')

fig, axs = plt.subplots(1, 2, figsize=(8, 3))
axs[0].plot(np.degrees(theta), Rs, label='R_s')
axs[0].plot(np.degrees(theta), Rp, label='R_p')
axs[0].axvline(np.degrees(theta_B), color='k', linestyle=':', label='Brewster')
axs[0].set_title('air -> glass')
axs[1].plot(np.degrees(theta), Rs2, label='R_s')
axs[1].plot(np.degrees(theta), Rp2, label='R_p')
axs[1].axvline(np.degrees(theta_c), color='r', linestyle='--', label='critical angle')
axs[1].set_title('glass -> air')
for ax in axs:
    ax.set_xlabel('incidence angle (deg)')
    ax.set_ylabel('reflectance')
plot_nicely()
plt.show()
```
