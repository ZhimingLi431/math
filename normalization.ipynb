最小最大值归一化 Min-Max Normalization
$$
x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}
$$


均值归一化 Mean Normalization
$$
x_{norm} = \frac{x - \mu}{x_{max} - x_{min}}
$$

稳健归一化 Robust Normalization
$$
x_{norm} = \frac{x - x_{median}}{x_{IQR}}
$$
$$
x_{IQR} = Q3 - Q1
$$
$$
Q3 = [x_{median}, x_{\max}]_{median}
$$
$$
Q1 = [x_{\min}, x_{median}]_{median}
$$
$$
\text{IQR: Interquartile Range}
$$
