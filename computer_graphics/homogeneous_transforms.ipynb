<a id="homogeneous"></a>

$$
\text{齐次坐标与变换矩阵 (Homogeneous Coordinates and Transformation Matrices)} \\ \quad \\
\text{线性变换 (Linear Transformation) 的定义: } T(a u + b v) = a T(u) + b T(v) \\
\text{取 } a = b = 0 \text{: } T(0) = 0 \text{, 线性变换一定把原点映射到原点} \\ \quad \\
\text{平移 (Translation) } T(p) = p + t \text{: } T(0) = t \neq 0 \text{, 所以平移不是线性变换, 不能写成 2x2 矩阵乘法} \\ \quad \\
\text{解决: 增加一个维度, 点 } (x, y) \to (x, y, 1) \\
\begin{bmatrix} 1 & 0 & t_x \\ 0 & 1 & t_y \\ 0 & 0 & 1 \end{bmatrix}
\begin{bmatrix} x \\ y \\ 1 \end{bmatrix} = \begin{bmatrix} x + t_x \\ y + t_y \\ 1 \end{bmatrix} \\
\text{在 3 维空间里, 这是一个剪切 (shear), 是线性的; 它在 } w = 1 \text{ 这个平面上的效果正好是平移} \\ \quad \\
\text{仿射变换 (Affine Transformation) = 线性变换 + 平移: } p' = A p + t \iff
\begin{bmatrix} p' \\ 1 \end{bmatrix} = \begin{bmatrix} A & t \\ 0 & 1 \end{bmatrix} \begin{bmatrix} p \\ 1 \end{bmatrix} \\ \quad \\
\text{点 (Point) 和向量 (Vector) 的区别: 点用 } w = 1 \text{, 方向向量用 } w = 0 \\
\text{两个点相减: } (p_1, 1) - (p_2, 1) = (p_1 - p_2, 0) \text{ 得到向量; 向量的 w = 0, 平移那一列乘以 0, 所以方向不受平移影响}
$$

数学基础: [矩阵乘法](../math_basics/linear_algebra.ipynb#matrix-multiplication)

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(suppress=True, precision=4)

def plot_nicely():
    for ax in plt.gcf().get_axes():
        ax.grid(True, which="both", alpha=0.3)
        if ax.get_legend_handles_labels()[0]:
            ax.legend(fontsize='small')
    plt.tight_layout()

def translate2d(tx, ty):
    return np.array([[1.0, 0, tx], [0, 1, ty], [0, 0, 1]])

# 平移不是线性的: T(0) != 0, T(u + v) != T(u) + T(v)
shift = lambda p: p + np.array([2.0, 1.0])
u, v = np.array([1.0, 0.0]), np.array([0.0, 3.0])
print(f'T(0) = {shift(np.zeros(2))}, T(u+v) = {shift(u + v)}, T(u)+T(v) = {shift(u) + shift(v)}')

T = translate2d(2, 1)
point = np.array([1.0, 0.0, 1.0])       # w = 1
direction = np.array([1.0, 0.0, 0.0])   # w = 0
print(f'点   (1, 0, 1) 平移后 = {T @ point}')
print(f'向量 (1, 0, 0) 平移后 = {T @ direction} (方向不受平移影响)')
p1, p2 = np.array([3.0, 4.0, 1.0]), np.array([1.0, 1.0, 1.0])
print(f'先相减再平移 = {T @ (p1 - p2)}, 先平移再相减 = {T @ p1 - T @ p2} (相同)')
```

<a id="basic-2d"></a>

$$
\text{基本 2D 变换: 矩阵的列 = 基向量变换后的位置} \\
A e_1 = A \begin{bmatrix} 1 \\ 0 \end{bmatrix} = \text{A 的第 1 列}, \quad A e_2 = \text{A 的第 2 列} \\
\text{任意向量 } p = x e_1 + y e_2 \text{, 由线性性 } A p = x (A e_1) + y (A e_2) \text{, 所以只要知道基向量去了哪里就能写出矩阵} \\ \quad \\
\text{1. 缩放 (Scale): } e_1 \to (s_x, 0), \ e_2 \to (0, s_y) \implies S = \begin{bmatrix} s_x & 0 \\ 0 & s_y \end{bmatrix} \\ \quad \\
\text{2. 旋转 (Rotation) 逆时针 } \theta \text{: 单位圆上 } e_1 \to (\cos\theta, \sin\theta), \ e_2 \text{ (在 90° 位置) } \to (\cos(90° + \theta), \sin(90° + \theta)) = (-\sin\theta, \cos\theta) \\
R(\theta) = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix} \\ \quad \\
\text{3. 剪切 (Shear) 沿 x 方向: } x' = x + k y \text{, 即 } e_1 \text{ 不动, } e_2 \to (k, 1) \implies H = \begin{bmatrix} 1 & k \\ 0 & 1 \end{bmatrix} \\ \quad \\
\text{4. 反射 (Reflection) 关于 x 轴: } e_1 \to (1, 0), \ e_2 \to (0, -1) \implies F = \begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix} \\ \quad \\
\text{行列式 = 面积缩放倍数 (带符号): } \det S = s_x s_y, \ \det R = \cos^2\theta + \sin^2\theta = 1, \ \det H = 1 \text{ (剪切不改变面积)}, \ \det F = -1 \text{ (翻转方向)}
$$

数学基础: [行列式与面积缩放](../math_basics/linear_algebra.ipynb#determinant)

```python
def scale2d(sx, sy):
    return np.array([[sx, 0, 0], [0, sy, 0], [0, 0, 1.0]])

def rotate2d(theta):
    c, s = np.cos(theta), np.sin(theta)
    return np.array([[c, -s, 0], [s, c, 0], [0, 0, 1.0]])

def shear2d(k):
    return np.array([[1.0, k, 0], [0, 1, 0], [0, 0, 1]])

def reflect_x():
    return np.diag([1.0, -1.0, 1.0])

# 一个 "F" 形状的多边形 (不对称, 能看出旋转和翻转)
shape = np.array([[0, 0], [0, 3], [2, 3], [2, 2.5], [0.5, 2.5], [0.5, 1.8], [1.5, 1.8], [1.5, 1.3], [0.5, 1.3], [0.5, 0], [0, 0]])
shape_h = np.column_stack([shape, np.ones(len(shape))]).T     # 3 x N, 每一列是一个点

def polygon_area(pts):
    # 鞋带公式 (Shoelace Formula), 带符号
    x, y = pts[:, 0], pts[:, 1]
    return 0.5 * np.sum(x[:-1] * y[1:] - x[1:] * y[:-1])

transforms = {
    'scale (1.5, 0.5)': scale2d(1.5, 0.5),
    'rotate 30 deg': rotate2d(np.deg2rad(30)),
    'shear k=0.6': shear2d(0.6),
    'reflect x': reflect_x(),
}
area0 = polygon_area(shape)
fig, axs = plt.subplots(1, 4, figsize=(12, 3))
for ax, (name, M) in zip(axs, transforms.items()):
    out = (M @ shape_h)[:2].T
    ratio = polygon_area(out) / area0
    print(f'{name:<18}: 面积比 = {ratio:+.4f}, det = {np.linalg.det(M[:2, :2]):+.4f}')
    ax.plot(shape[:, 0], shape[:, 1], 'k--', label='original')
    ax.plot(out[:, 0], out[:, 1], 'r', label=name)
    ax.set_aspect('equal')
plot_nicely()
plt.show()

# 列 = 基向量的像
R = rotate2d(np.deg2rad(30))[:2, :2]
print(f'R e1 = {R @ [1, 0]}, R 的第 1 列 = {R[:, 0]}; R e2 = {R @ [0, 1]}, R 的第 2 列 = {R[:, 1]}')
```

<a id="composition"></a>

$$
\text{变换的组合 (Composition): 矩阵相乘, 从右往左依次作用} \\
M = T R S \implies M p = T(R(S p)) \quad (\text{先缩放, 再旋转, 最后平移, 这是建模中最常用的 TRS 顺序}) \\ \quad \\
\text{矩阵乘法不满足交换律, 顺序不同结果不同. 例如 TR 和 RT:} \\
T R \begin{bmatrix} p \\ 1 \end{bmatrix} = R p + t \quad (\text{先绕原点转, 再平移}) \\
R T \begin{bmatrix} p \\ 1 \end{bmatrix} = R (p + t) = R p + R t \quad (\text{先平移, 再绕原点转, 平移量也被转了}) \\ \quad \\
\text{绕任意点 c 旋转: 先把 c 移到原点, 旋转, 再移回去} \\
M = T(c) \, R(\theta) \, T(-c) \implies M p = R(p - c) + c \\
\text{检查: } M c = R \cdot 0 + c = c \text{, 旋转中心不动}
$$

```python
theta = np.deg2rad(90)
Tm, Rm = translate2d(3, 0), rotate2d(theta)
TR = (Tm @ Rm @ shape_h)[:2].T
RT = (Rm @ Tm @ shape_h)[:2].T
print(f'TR == RT ? {np.allclose(TR, RT)}')
print(f'TR 的平移列 = {(Tm @ Rm)[:2, 2]}, RT 的平移列 = {(Rm @ Tm)[:2, 2]} (= R t)')

center = np.array([1.0, 1.5])
M_center = translate2d(*center) @ Rm @ translate2d(*(-center))
print(f'绕点 c 旋转后 c 的位置 = {(M_center @ np.append(center, 1))[:2]}')
p_test = np.array([2.0, 0.5])
print(f'M p = {(M_center @ np.append(p_test, 1))[:2]}, R(p - c) + c = {Rm[:2, :2] @ (p_test - center) + center}')

fig, axs = plt.subplots(1, 2, figsize=(8, 3))
axs[0].plot(shape[:, 0], shape[:, 1], 'k--', label='original')
axs[0].plot(TR[:, 0], TR[:, 1], 'r', label='T R (rotate, then translate)')
axs[0].plot(RT[:, 0], RT[:, 1], 'b', label='R T (translate, then rotate)')
rc = (M_center @ shape_h)[:2].T
axs[1].plot(shape[:, 0], shape[:, 1], 'k--', label='original')
axs[1].plot(rc[:, 0], rc[:, 1], 'g', label='rotate about c')
axs[1].plot(*center, 'ko', label='c')
for ax in axs:
    ax.set_aspect('equal')
plot_nicely()
plt.show()
```

<a id="inverse-transform"></a>

$$
\text{逆变换 (Inverse Transformation)} \\
T(t)^{-1} = T(-t), \quad S(s_x, s_y)^{-1} = S(\tfrac{1}{s_x}, \tfrac{1}{s_y}), \quad R(\theta)^{-1} = R(-\theta) = R(\theta)^T \text{ (正交矩阵)} \\ \quad \\
\text{组合的逆: } (AB)^{-1} = B^{-1} A^{-1} \quad (\text{验证: } AB \cdot B^{-1} A^{-1} = A I A^{-1} = I) \\
(T R S)^{-1} = S^{-1} R^{-1} T^{-1} \quad (\text{穿衣服的逆过程是倒过来脱}) \\ \quad \\
\text{一般仿射变换的逆: 解 } y = A x + t \implies x = A^{-1}(y - t) = A^{-1} y - A^{-1} t \\
\begin{bmatrix} A & t \\ 0 & 1 \end{bmatrix}^{-1} = \begin{bmatrix} A^{-1} & -A^{-1} t \\ 0 & 1 \end{bmatrix} \\
\text{只需要求 A 的逆 (2x2 或 3x3), 不用对整个 4x4 求逆; A 是旋转时 } A^{-1} = A^T \text{, 更简单}
$$

数学基础: [矩阵的逆](../math_basics/linear_algebra.ipynb#inverse) · [正交矩阵](../math_basics/linear_algebra.ipynb#orthogonal)

```python
S_, R_, T_ = scale2d(2, 0.5), rotate2d(0.7), translate2d(-1, 4)
M = T_ @ R_ @ S_
M_inv_reverse = np.linalg.inv(S_) @ R_.T @ translate2d(1, -4)
print(f'(TRS)^-1 == S^-1 R^T T(-t) ? {np.allclose(M_inv_reverse, np.linalg.inv(M))}')
print(f'错误顺序 T^-1 R^-1 S^-1 == 逆 ? {np.allclose(translate2d(1, -4) @ R_.T @ np.linalg.inv(S_), np.linalg.inv(M))}')

def affine_inverse(M):
    A, t = M[:-1, :-1], M[:-1, -1]
    A_inv = np.linalg.inv(A)
    out = np.eye(len(M))
    out[:-1, :-1] = A_inv
    out[:-1, -1] = -A_inv @ t
    return out

print(f'分块公式 == np.linalg.inv ? {np.allclose(affine_inverse(M), np.linalg.inv(M))}')
print(f'M^-1 M 作用在形状上回到原样 ? {np.allclose(affine_inverse(M) @ M @ shape_h, shape_h)}')
```

<a id="transform-3d"></a>

$$
\text{3D 变换: 4x4 齐次矩阵, 点 } (x, y, z, 1) \\
T = \begin{bmatrix} 1 & 0 & 0 & t_x \\ 0 & 1 & 0 & t_y \\ 0 & 0 & 1 & t_z \\ 0 & 0 & 0 & 1 \end{bmatrix}, \quad
S = \begin{bmatrix} s_x & 0 & 0 & 0 \\ 0 & s_y & 0 & 0 \\ 0 & 0 & s_z & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix} \\ \quad \\
\text{绕坐标轴旋转 = 在另外两个坐标组成的平面里做 2D 旋转, 旋转轴上的坐标不变} \\
\text{绕 z 轴: (x, y) 平面, x → y 方向转} \implies R_z = \begin{bmatrix} \cos\theta & -\sin\theta & 0 \\ \sin\theta & \cos\theta & 0 \\ 0 & 0 & 1 \end{bmatrix} \\
\text{绕 x 轴: (y, z) 平面, y → z 方向转} \implies R_x = \begin{bmatrix} 1 & 0 & 0 \\ 0 & \cos\theta & -\sin\theta \\ 0 & \sin\theta & \cos\theta \end{bmatrix} \\
\text{绕 y 轴: 按右手系的循环顺序 } x \to y \to z \to x \text{, 这个平面是 (z, x), z → x 方向转} \\
z' = z \cos\theta - x \sin\theta, \quad x' = z \sin\theta + x \cos\theta \implies R_y = \begin{bmatrix} \cos\theta & 0 & \sin\theta \\ 0 & 1 & 0 \\ -\sin\theta & 0 & \cos\theta \end{bmatrix} \\
\text{(所以 } R_y \text{ 里 sin 的符号和另外两个看起来是反的)} \\
\text{详细推导 (任意轴旋转, 四元数) 见 } \text{rotation\_and\_quaternion}
$$

```python
from scipy.spatial.transform import Rotation

def Rx(t):
    c, s = np.cos(t), np.sin(t)
    return np.array([[1, 0, 0], [0, c, -s], [0, s, c]])

def Ry(t):
    c, s = np.cos(t), np.sin(t)
    return np.array([[c, 0, s], [0, 1, 0], [-s, 0, c]])

def Rz(t):
    c, s = np.cos(t), np.sin(t)
    return np.array([[c, -s, 0], [s, c, 0], [0, 0, 1]])

ang = 0.4
for name, Rm3, axis in [('Rx', Rx(ang), [1, 0, 0]), ('Ry', Ry(ang), [0, 1, 0]), ('Rz', Rz(ang), [0, 0, 1])]:
    ref = Rotation.from_rotvec(ang * np.array(axis)).as_matrix()
    print(f'{name} == scipy Rotation ? {np.allclose(Rm3, ref)}')

# 右手系的循环: 绕 y 轴转 90 度, z 轴转到 x 轴
print(f'Ry(90 deg) @ z = {Ry(np.pi / 2) @ [0, 0, 1]} (z -> x)')
print(f'Rz(90 deg) @ x = {Rz(np.pi / 2) @ [1, 0, 0]} (x -> y), Rx(90 deg) @ y = {Rx(np.pi / 2) @ [0, 1, 0]} (y -> z)')
```

<a id="normal-transform"></a>

$$
\text{法向量的变换 (Transforming Normals): 为什么要用 } (M^{-1})^T \\ \quad \\
\text{法向量 n 垂直于表面上的切向量 (tangent) t: } n^T t = 0 \\
\text{切向量是表面上两个点的差, 所以和普通向量一样变换: } t' = M t \text{ (M 是左上角的 3x3 部分)} \\
\text{要找矩阵 G, 使得 } n' = G n \text{ 仍然垂直于 } t' \text{:} \\
n'^T t' = (G n)^T (M t) = n^T G^T M t \\
\text{如果 } G^T M = I \text{, 就有 } n'^T t' = n^T t = 0 \implies G^T = M^{-1} \implies G = (M^{-1})^T \\ \quad \\
\text{特殊情况:} \\
\text{1. M 是旋转 (正交矩阵): } (M^{-1})^T = (M^T)^T = M \text{, 法向量和普通向量一样变换} \\
\text{2. M 是均匀缩放 sI: } (M^{-1})^T = \frac{1}{s} I \text{, 只差一个长度, 归一化以后一样} \\
\text{3. M 含有非均匀缩放 / 剪切: 直接用 M 变换法向量会出错, 光照会算错}
$$

数学基础: [矩阵的逆](../math_basics/linear_algebra.ipynb#inverse) · [转置 (AB)^T = B^T A^T](../math_basics/linear_algebra.ipynb#transpose) · [正交矩阵](../math_basics/linear_algebra.ipynb#orthogonal)

```python
# 单位圆: 点 (cos a, sin a), 切向量 (-sin a, cos a), 法向量 (cos a, sin a)
a = np.linspace(0, 2 * np.pi, 16, endpoint=False)
circle_pts = np.column_stack([np.cos(a), np.sin(a)])
tangents = np.column_stack([-np.sin(a), np.cos(a)])
normals = circle_pts.copy()

M2 = np.array([[2.0, 0.0], [0.0, 0.5]])        # 非均匀缩放, 圆 -> 椭圆
ellipse_pts = circle_pts @ M2.T
t_new = tangents @ M2.T
n_wrong = normals @ M2.T
n_right = normals @ np.linalg.inv(M2)          # 行向量形式: (M^-1)^T n  <=>  n^T M^-1

dot_wrong = np.einsum('ij,ij->i', n_wrong, t_new)
dot_right = np.einsum('ij,ij->i', n_right, t_new)
print(f'用 M 变换法向量: n . t 最大值 = {np.max(np.abs(dot_wrong)):.4f} (不垂直)')
print(f'用 (M^-1)^T 变换: n . t 最大值 = {np.max(np.abs(dot_right)):.2e} (垂直)')

# 独立验证: 椭圆 x^2/4 + y^2/0.25 = 1 的法向量 = 隐函数的梯度 (2x/4, 2y/0.25) = (x/2, 8y) (梯度垂直于等高线)
grad_ellipse = np.column_stack([ellipse_pts[:, 0] / 2, 8 * ellipse_pts[:, 1]])
unit = lambda v: v / np.linalg.norm(v, axis=1, keepdims=True)
print(f'(M^-1)^T n 的方向 == 椭圆方程的梯度方向 ? {np.allclose(unit(n_right), unit(grad_ellipse))}')

# 旋转矩阵: (R^-1)^T == R
R3 = Rz(0.8) @ Rx(-0.3)
print(f'旋转矩阵 (R^-1)^T == R ? {np.allclose(np.linalg.inv(R3).T, R3)}')

plt.figure(figsize=(4, 3))
plt.plot(*np.vstack([ellipse_pts, ellipse_pts[:1]]).T, 'k')
s = 0.4
for p_, nw, nr in zip(ellipse_pts, unit(n_wrong), unit(n_right)):
    plt.plot([p_[0], p_[0] + s * nw[0]], [p_[1], p_[1] + s * nw[1]], 'r')
    plt.plot([p_[0], p_[0] + s * nr[0]], [p_[1], p_[1] + s * nr[1]], 'g')
plt.plot([], [], 'r', label='M n (wrong)')
plt.plot([], [], 'g', label='(M^-1)^T n (correct)')
plt.gca().set_aspect('equal')
plot_nicely()
plt.show()
```
