<a id="gravity"></a>

$$
\text{万有引力 (Universal Gravitation) 与轨道运动} \\
F = -\frac{G M m}{r^2} \hat{r} = -\frac{G M m}{r^3} \vec{r} \quad (\text{指向太阳, 大小和距离平方成反比}) \\
\text{牛顿第二定律: } \ddot{\vec{r}} = -\frac{G M}{r^3} \vec{r} \quad (\text{m 约掉了, 轨道和行星质量无关}) \\
\text{单位: 距离用天文单位 AU, 时间用年, 则地球轨道 (圆, r = 1, T = 1) 满足 } \frac{v^2}{r} = \frac{GM}{r^2}, \ v = 2\pi \implies GM = 4\pi^2 \\ \quad \\
\text{1. 角动量守恒 (Conservation of Angular Momentum): } \vec{L} = \vec{r} \times m\vec{v} \\
\frac{d\vec{L}}{dt} = \underbrace{\dot{\vec{r}} \times m\vec{v}}_{\vec{v} \times m\vec{v} = 0} + \vec{r} \times m\dot{\vec{v}} = \vec{r} \times \vec{F} = \vec{r} \times \left(-\frac{GMm}{r^3}\vec{r}\right) = 0 \\
\text{(乘积法则对叉乘也成立; 平行向量的叉乘为 0, 所有中心力 (central force) 都守恒角动量)} \\
\vec{L} \text{ 方向不变} \implies \vec{r} \text{ 始终垂直于固定的 } \vec{L} \implies \text{运动在一个平面内} \\ \quad \\
\text{2. 开普勒第二定律 (Kepler's 2nd Law): 相等时间扫过相等面积} \\
dt \text{ 时间内扫过的三角形面积 } dA = \frac{1}{2}|\vec{r} \times d\vec{r}| = \frac{1}{2}|\vec{r} \times \vec{v}| \, dt \implies \frac{dA}{dt} = \frac{|\vec{L}|}{2m} = \text{常数} \\ \quad \\
\text{3. 能量守恒: } E = \frac{1}{2} m v^2 - \frac{GMm}{r} \\
\frac{dE}{dt} = m \vec{v} \cdot \dot{\vec{v}} + \frac{GMm}{r^2}\dot{r} = \vec{v} \cdot \vec{F} + \frac{GMm}{r^2}\dot{r} = -\frac{GMm}{r^3}\vec{r} \cdot \vec{v} + \frac{GMm}{r^3} \vec{r} \cdot \vec{v} = 0 \\
\text{(用到 } r^2 = \vec{r} \cdot \vec{r} \implies 2r\dot{r} = 2\vec{r} \cdot \vec{v} \implies \dot{r} = \frac{\vec{r} \cdot \vec{v}}{r} \text{)}
$$

数学基础: [求导法则 (乘积法则)](../math_basics/calculus.ipynb#derivative-rules) · [高阶 ODE 转一阶方程组](../math_basics/calculus.ipynb#ode-system)

<a id="effective-potential"></a>

$$
\text{有效势能 (Effective Potential): 把二维运动化成一维的径向运动} \\
\text{极坐标 } \vec{r} = r(\cos\theta, \sin\theta) \text{, 速度分解为径向 } \dot{r} \text{ 和切向 } r\dot{\theta} \text{: } v^2 = \dot{r}^2 + r^2\dot{\theta}^2 \\
\text{角动量 } L = m r^2 \dot{\theta} \implies \dot{\theta} = \frac{L}{m r^2} \\
E = \frac{1}{2} m \dot{r}^2 + \underbrace{\frac{L^2}{2 m r^2} - \frac{G M m}{r}}_{U_{\text{eff}}(r)} \\
\text{离心势 (centrifugal) } \frac{L^2}{2mr^2} \text{ 在 r 很小时主导, 阻止行星落入太阳} \\ \quad \\
\text{因为 } \frac{1}{2} m \dot{r}^2 \geq 0 \text{, 所以运动只能在 } U_{\text{eff}}(r) \leq E \text{ 的区域; } \dot{r} = 0 \text{ 的点是近日点 / 远日点 (turning points)} \\
E < 0 \text{: 束缚轨道 (椭圆); } E = 0 \text{: 抛物线; } E > 0 \text{: 双曲线 (逃逸)} \\ \quad \\
\text{圆轨道 = } U_{\text{eff}} \text{ 的最小值: } \frac{dU_{\text{eff}}}{dr} = -\frac{L^2}{m r^3} + \frac{GMm}{r^2} = 0 \implies r_c = \frac{L^2}{G M m^2} \\ \quad \\
\text{轨道方程 (Binet 方程): 令 } u = \frac{1}{r}, \ h = \frac{L}{m} = r^2\dot{\theta} \\
\dot{r} = \frac{dr}{d\theta}\dot{\theta} = -\frac{1}{u^2}\frac{du}{d\theta} \cdot h u^2 = -h \frac{du}{d\theta}, \quad \ddot{r} = -h \frac{d^2 u}{d\theta^2}\dot{\theta} = -h^2 u^2 \frac{d^2 u}{d\theta^2} \\
\text{径向方程 } \ddot{r} - r\dot{\theta}^2 = -\frac{GM}{r^2} \text{ (极坐标下的径向加速度) 代入:} \\
-h^2 u^2 u'' - h^2 u^3 = -GM u^2 \implies u'' + u = \frac{GM}{h^2} \\
\text{(和简谐振动同一类方程, 加一个常数项)} \implies u = \frac{GM}{h^2}(1 + e\cos\theta) \implies r = \frac{p}{1 + e\cos\theta}, \ p = \frac{h^2}{GM} \\
\text{这是圆锥曲线 (conic section) 的极坐标方程, e 是离心率: } 0 \leq e < 1 \text{ 为椭圆 (开普勒第一定律)}
$$

数学基础: [极值的一阶条件](../math_basics/calculus.ipynb#extremum) · [链式法则](../math_basics/calculus.ipynb#derivative-rules) · 应用: [简谐振动](harmonic_oscillator.ipynb#simple-harmonic)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

np.set_printoptions(linewidth=np.inf)
np.set_printoptions(suppress=True)

def plot_nicely():
    for ax in plt.gcf().get_axes():
        ax.grid(True, which="both", alpha=0.3)
        if ax.get_legend_handles_labels()[0]:
            ax.legend(fontsize='small')
    plt.tight_layout()

GM = 4 * np.pi**2

def accel(r):
    return -GM * r / np.linalg.norm(r)**3

def rhs(t, s):
    return np.concatenate([s[2:], accel(s[:2])])

# 近日点出发: r = (r_p, 0), v = (0, v_p)
a_orbit, e_orbit = 1.0, 0.6
r_p = a_orbit * (1 - e_orbit)
v_p = np.sqrt(GM * (1 + e_orbit) / r_p)   # 近日点速度 (vis-viva 公式, 由能量守恒 + 角动量守恒推出)
s0 = np.array([r_p, 0.0, 0.0, v_p])

t_eval = np.linspace(0, 1, 20001)   # 一个周期 (T = a^1.5 = 1 年)
sol = solve_ivp(rhs, [0, 1], s0, t_eval=t_eval, rtol=1e-12, atol=1e-12)
r_vec, v_vec = sol.y[:2].T, sol.y[2:].T
r_norm = np.linalg.norm(r_vec, axis=1)

L = r_vec[:, 0] * v_vec[:, 1] - r_vec[:, 1] * v_vec[:, 0]   # 单位质量的角动量 (z 分量)
E = 0.5 * np.sum(v_vec**2, axis=1) - GM / r_norm
print(f'角动量: 最小 = {L.min():.10f}, 最大 = {L.max():.10f}')
print(f'能量:   最小 = {E.min():.10f}, 最大 = {E.max():.10f}, 理论 -GM / (2a) = {-GM / (2 * a_orbit):.10f}')

# 开普勒第二定律: 每 0.1 年扫过的面积 (三角形面积 1/2 |r x dr| 求和)
cross = r_vec[:-1, 0] * r_vec[1:, 1] - r_vec[:-1, 1] * r_vec[1:, 0]
areas = (0.5 * cross).reshape(10, 2000).sum(axis=1)
print(f'每 0.1 年扫过的面积 = {np.round(areas, 6)}, L / 2 * 0.1 = {L[0] / 2 * 0.1:.6f}')

# 轨道方程 r = p / (1 + e cos theta)
theta = np.arctan2(r_vec[:, 1], r_vec[:, 0])
p = L[0]**2 / GM
print(f'p = h^2 / GM = {p:.6f}, a(1 - e^2) = {a_orbit * (1 - e_orbit**2):.6f}; 数值轨道 == p / (1 + e cos theta)? {np.allclose(r_norm, p / (1 + e_orbit * np.cos(theta)), atol=1e-8)}')

# 有效势能与转折点
r_grid = np.linspace(0.15, 3, 40001)
U_eff = L[0]**2 / (2 * r_grid**2) - GM / r_grid
r_c = L[0]**2 / GM
print(f'U_eff 最小值: 网格 r = {r_grid[np.argmin(U_eff)]:.4f}, 公式 r_c = L^2 / GM = {r_c:.4f}; 近日点 = {r_norm.min():.4f}, 远日点 = {r_norm.max():.4f} (a(1 +- e) = {r_p:.4f}, {a_orbit * (1 + e_orbit):.4f})')

fig, axs = plt.subplots(1, 2, figsize=(8, 3))
axs[0].plot(r_vec[:, 0], r_vec[:, 1], label=f'orbit e={e_orbit}')
axs[0].plot(0, 0, 'y*', markersize=12, label='sun')
for k in range(0, 20000, 4000):   # 每 0.2 年扫过的扇形
    axs[0].fill([0, *r_vec[k:k + 2001:50, 0], 0], [0, *r_vec[k:k + 2001:50, 1], 0], alpha=0.3)
axs[0].set_aspect('equal')
axs[0].set_title('equal areas in equal times')
axs[1].plot(r_grid, U_eff, label='U_eff(r)')
axs[1].axhline(E[0], color='r', linestyle='--', label='E')
axs[1].axvline(r_c, color='gray', linestyle=':', label='circular r_c')
axs[1].set_ylim(-60, 20)
axs[1].set_xlabel('r (AU)')
plot_nicely()
plt.show()
```

<a id="verlet"></a>

$$
\text{辛积分 (Symplectic Integrator): 速度 Verlet (Velocity Verlet) / 蛙跳法 (Leapfrog)} \\ \quad \\
\text{推导: 位置用二阶泰勒展开, 速度用梯形法积分加速度} \\
r(t + h) = r(t) + h \, v(t) + \frac{h^2}{2} a(t) + O(h^3) \\
v(t + h) = v(t) + \int_t^{t+h} a(s) \, ds \approx v(t) + \frac{h}{2}\left[a(t) + a(t + h)\right] \quad (\text{梯形法}) \\
\text{算法: 1. } r_{n+1} = r_n + h v_n + \frac{h^2}{2} a_n \quad \text{2. } a_{n+1} = a(r_{n+1}) \quad \text{3. } v_{n+1} = v_n + \frac{h}{2}(a_n + a_{n+1}) \\
\text{全局误差 } O(h^2) \text{, 每步只需要算一次加速度 (RK4 要 4 次)} \\ \quad \\
\text{为什么长期模拟用它: 时间可逆 (把 h 换成 -h 可以精确地走回去), 并且保持相空间面积 (辛结构)} \\
\text{结果: 能量误差在一个固定范围内振荡, 不会累积; RK4 每一步精度高, 但能量误差单调漂移, 时间长了轨道会慢慢衰减或扩张} \\
\text{(和 calculus 里欧拉法让谐振子能量指数增长是同一个问题)}
$$

数学基础: [泰勒展开](../math_basics/calculus.ipynb#taylor) · [梯形法](../math_basics/calculus.ipynb#numerical-integral) · [Euler / RK4](../math_basics/calculus.ipynb#euler-rk4)

```python
def velocity_verlet(r0, v0, h, n_steps):
    r, v = np.array(r0, float), np.array(v0, float)
    a = accel(r)
    rs, vs = [r.copy()], [v.copy()]
    for _ in range(n_steps):
        r = r + h * v + 0.5 * h**2 * a
        a_new = accel(r)
        v = v + 0.5 * h * (a + a_new)
        a = a_new
        rs.append(r.copy()); vs.append(v.copy())
    return np.array(rs), np.array(vs)

def rk4_orbit(r0, v0, h, n_steps):
    s = np.concatenate([r0, v0]).astype(float)
    f = lambda s: np.concatenate([s[2:], accel(s[:2])])
    out = [s.copy()]
    for _ in range(n_steps):
        k1 = f(s); k2 = f(s + h / 2 * k1); k3 = f(s + h / 2 * k2); k4 = f(s + h * k3)
        s = s + h / 6 * (k1 + 2 * k2 + 2 * k3 + k4)
        out.append(s.copy())
    out = np.array(out)
    return out[:, :2], out[:, 2:]

energy = lambda r, v: 0.5 * np.sum(v**2, axis=1) - GM / np.linalg.norm(r, axis=1)

# 公平比较: 计算量相同 (RK4 每步算 4 次加速度, Verlet 1 次), 所以 Verlet 的步长取 RK4 的 1/4
years = 200
h_rk = 0.01
h_ver = h_rk / 4
r_ver, v_ver = velocity_verlet(s0[:2], s0[2:], h_ver, int(years / h_ver))
r_rk, v_rk = rk4_orbit(s0[:2], s0[2:], h_rk, int(years / h_rk))
E0 = energy(s0[None, :2], s0[None, 2:])[0]
dE_ver = energy(r_ver, v_ver) / E0 - 1
dE_rk = energy(r_rk, v_rk) / E0 - 1
t_ver = np.arange(len(dE_ver)) * h_ver
t_rk = np.arange(len(dE_rk)) * h_rk

# 时间可逆: Verlet 走 n 步再用 -h 走回来
r_back, v_back = velocity_verlet(r_ver[1000], v_ver[1000], -h_ver, 1000)
print(f'Verlet 时间可逆: 走 1000 步再倒回 1000 步, 和起点的差 = {np.max(np.abs(r_back[-1] - s0[:2])):.2e}')

for yr in [10, 50, 100, 200]:
    kv, kr = int(round(yr / h_ver)), int(round(yr / h_rk))
    print(f't = {yr:>3} 年: Verlet 能量相对误差 = {dE_ver[kv]:+.2e} (至今最大 |误差| {np.max(np.abs(dE_ver[:kv + 1])):.2e}), '
          f'RK4 = {dE_rk[kr]:+.2e} (至今最大 {np.max(np.abs(dE_rk[:kr + 1])):.2e})')
print('Verlet 的误差在一个固定范围内振荡; RK4 的误差随时间线性累积, 时间越长差距越大')

plt.figure(figsize=(4, 3))
plt.plot(t_ver, dE_ver, label=f'velocity Verlet (h={h_ver})', alpha=0.7)
plt.plot(t_rk, dE_rk, label=f'RK4 (h={h_rk})')
plt.xlabel('t (years)')
plt.ylabel('relative energy error')
plot_nicely()
plt.show()
```

<a id="kepler-third"></a>

$$
\text{开普勒第三定律 (Kepler's 3rd Law): } T^2 \propto a^3 \\ \quad \\
\text{1. 椭圆面积 } \pi a b \text{, 由第二定律扫面积的速度恒为 } \frac{h}{2} \implies T = \frac{\pi a b}{h / 2} = \frac{2\pi a b}{h} \\
\text{2. 椭圆的几何关系: 半通径 } p = \frac{b^2}{a} \text{, 而轨道方程给出 } p = \frac{h^2}{GM} \implies h^2 = \frac{GM b^2}{a} \\
\text{3. } T^2 = \frac{4\pi^2 a^2 b^2}{h^2} = \frac{4\pi^2 a^2 b^2 \cdot a}{GM b^2} = \frac{4\pi^2}{GM} a^3 \\
\text{在 AU-年 单位下 } GM = 4\pi^2 \implies T^2 = a^3 \text{ (地球 a = 1, T = 1)} \\ \quad \\
\text{半长轴由能量决定: 近日点和远日点能量相等 + 角动量相等, 可以推出 } E = -\frac{GM}{2a} \text{ (单位质量)} \\
\text{(上面数值验证过能量 = } -GM / (2a) \text{), 再由 } E = \frac{v^2}{2} - \frac{GM}{r} \text{ 得 vis-viva 公式 } v^2 = GM\left(\frac{2}{r} - \frac{1}{a}\right)
$$

```python
def measure_period(r0, v0):
    # theta 从 0 出发, 再次从下向上穿过 x 轴正半轴 (y 从负变正) 时为一个周期
    def cross(t, s):
        return s[1]
    cross.direction = 1
    sol = solve_ivp(rhs, [0, 30], np.concatenate([r0, v0]), events=cross, rtol=1e-12, atol=1e-12)
    events = sol.t_events[0]
    return events[events > 1e-6][0]   # t = 0 时 y = 0 也会被记成一次穿越, 跳过

rows = []
for r_start, speed_factor in [(0.4, 1.0), (1.0, 1.2), (1.5, 0.9), (2.0, 1.1), (5.2, 1.0)]:
    v_circ = np.sqrt(GM / r_start)
    r0, v0 = np.array([r_start, 0.0]), np.array([0.0, speed_factor * v_circ])
    E_spec = 0.5 * np.sum(v0**2) - GM / r_start
    a_semi = -GM / (2 * E_spec)
    T_num = measure_period(r0, v0)
    rows.append((a_semi, T_num))
    print(f'r0 = {r_start}, v0 = {speed_factor} v_circ: a = {a_semi:.4f}, T 数值 = {T_num:.6f}, a^1.5 = {a_semi**1.5:.6f}, T^2 / a^3 = {T_num**2 / a_semi**3:.8f}')

a_arr, T_arr = np.array(rows).T
slope = np.polyfit(np.log(a_arr), np.log(T_arr), 1)[0]
print(f'log T 对 log a 的斜率 = {slope:.6f} (理论 1.5)')
```
